## 1. Indentation — the "shape" of code

Most languages use braces `{ … }` to group code. Python uses **indentation**: lines that belong to the same block must start in the same column.

```
if score >= 80:
····print("Auto-approve")        ← 4 spaces in → inside the if
····send_email(user)             ← also inside
print("All done.")               ← back to column 0 → outside the if
```

Rules of thumb:

- The standard is **4 spaces** per level. Don't mix spaces and tabs.
- A "block-opening" line ends in `:` (e.g. `if x > 0:`, `for i in range(3):`, `def f():`).
- The block ends when indentation goes back to the previous level.

Most editors (including Colab) insert 4 spaces automatically when you press Tab after a `:` line.

````python
confidence = 0.92

if confidence > 0.8:
    print("✅ Auto-approve this answer")
    print(f"   (confidence = {confidence:.0%})")
print("After the if-block (always runs).")
````

> ⚠️ **IndentationError** is the single most common error new Python users see. If you see it, look at the column where each line *starts*. The usual culprit is mixed tabs and spaces.

## 2. `if / elif / else` — building a rules engine

The structure mirrors how product teams describe automation rules:

> *"If the model is very confident, auto-respond. Otherwise if it's somewhat confident, suggest a draft. Otherwise route to a human."*

````python
# A two-way decision: do we send or hold?
score = 0.62

if score >= 0.5:
    print("➡️  Send the automated reply")
else:
    print("🛑  Route to a human reviewer")
````

> 🔭 **Forward-reference — `def`.** The next cell wraps its logic in a **function**. Read `def name(arg):` as "define a reusable recipe called *name* that takes an input *arg*", and `return` as "hand a value back to whoever called it". We use small functions throughout this notebook because they keep the examples tidy — you'll get the *full* treatment (parameters, defaults, scope, docstrings) in **Notebook 5**. For now you only need to *read* them, not write them.

````python
# Multi-way decision — a triage rules engine for an AI support bot
def route_ticket(confidence):
    if confidence >= 0.90:
        return "auto-respond"
    elif confidence >= 0.70:
        return "draft-for-agent"
    elif confidence >= 0.40:
        return "human-only"
    else:
        return "needs-clarification"

# Try a few confidence scores
for c in [0.95, 0.78, 0.55, 0.30]:
    print(f"  confidence={c:.2f}  →  {route_ticket(c)}")
````

**How Python evaluates `elif`.** Python checks branches *in order* and runs the **first** one whose condition is true. The rest are skipped.

> ⚠️ **Order matters.** If you wrote `confidence >= 0.40` first, *every* high-confidence ticket would fall into the "human-only" branch. As a rule of thumb, in a tiered rule like this go from the **most restrictive** condition to the **least**.

> 🎯 **Intuition.** Think of `if/elif/else` as a fall-through staircase. As soon as a step is "true", you step off; lower steps are never checked.

### Truthiness — when "falsy" values trip you up

Python treats several values as **falsy** in a boolean context: `0`, `0.0`, `""` (empty string), `None`, and empty containers like `[]`, `{}`, `()`. Everything else is **truthy**.

This is convenient (`if api_response:` reads as "if we got something back"), but watch out: `if score:` is `True` even when `score = -10`, but `False` when `score = 0`.

````python
for value in [0, 1, -1, 0.0, "", "hi", None, [], [0], {}]:
    print(f"{repr(value):>6} → {'truthy' if value else 'falsy'}")
````

### 🔬 What actually happens when you write `if items:`

When you write `if items:`, it *looks* like Python checks `items` directly. It doesn't. Behind the scenes Python **converts whatever you give it into a real `True`/`False` by calling `bool()` on it.** These two lines are exactly equivalent:

```python
if items:          # what you write
if bool(items):    # what Python actually evaluates
```

So the condition of *every* `if`, `while`, `and`, `or`, and `not` is silently funnelled through `bool(...)`. That single rule is the whole story of "truthiness".

```text
   if items:
      │   Python rewrites the test
      ▼
   if bool(items):
      │   bool() asks the object: "are you empty / zero / nothing?"
      ▼
   ┌────────────────────────────┬────────────────────────────┐
   │  empty / zero / None       │  anything else             │
   │  → bool(...) is False      │  → bool(...) is True       │
   │  → the if-block is SKIPPED │  → the if-block RUNS       │
   └────────────────────────────┴────────────────────────────┘
```

A short, fixed list of values are **falsy** (their `bool()` is `False`). *Everything* else is **truthy**.


### The complete falsy list — and why each one is falsy

There are only a handful of falsy values. Memorise the *shape* of them — "empty or zero or nothing" — and you'll never need to memorise the literal list:

| Value | `bool(...)` | Why it's falsy |
|---|---|---|
| `False` | `False` | it literally *is* false |
| `None`  | `False` | "no value at all" |
| `0`     | `False` | the integer zero |
| `0.0`   | `False` | the float zero |
| `""`    | `False` | the **empty** string |
| `[]`    | `False` | the **empty** list |
| `{}`    | `False` | the **empty** dict |
| `set()` | `False` | the **empty** set |
| `()`    | `False` | the **empty** tuple |

Everything else is truthy — including ones that surprise beginners: `-1`, `0.0001`, `"0"` (a one-character string!), `" "` (a space), `[0]` (a list that *contains* a zero), and `[[]]` (a list that contains an empty list). The cell below proves the whole table at once:


````python
# PROOF: `if x:` is really `if bool(x):` — let's call bool() on every value ourselves.
falsy   = [False, None, 0, 0.0, "", [], {}, set(), ()]
truthy  = [True, -1, 0.0001, "0", " ", [0], [[]], {"k": 1}, {0}]

print("FALSY  (bool(x) is False):")
for x in falsy:
    print(f"   bool({repr(x):<7}) = {bool(x)}")

print("\nTRUTHY (bool(x) is True) — note the surprises:")
for x in truthy:
    print(f"   bool({repr(x):<7}) = {bool(x)}")

# And the equivalence itself, made explicit:
items = []
print("\nSame result either way:")
print("  'if items:'        →", "runs" if items       else "skipped")
print("  'if bool(items):'  →", "runs" if bool(items) else "skipped")
````

### The `if not items:` idiom — and the trap to avoid

Because empty containers are falsy, Python has an idiomatic way to ask *"is this empty / missing?"* — just write `if not items:`. It's shorter and reads like English:

```python
if not results:                  # ✅ Pythonic: "if there are no results"
    print("nothing came back")

if len(results) == 0:            # 😐 works, but verbose and less idiomatic
    print("nothing came back")
```

> ⚠️ **The truthiness trap.** `if x:` and `if x is not None:` are **not** the same test, and confusing them causes real bugs. Imagine a function that returns a *count*: a count of `0` is a perfectly valid answer, but `0` is falsy! If you write `if count:` you'll wrongly skip the zero case. When you specifically mean "was a value provided at all?", test `is None` explicitly — don't lean on truthiness.


````python
# ⚠️ Why `if x:` and `if x is not None:` are different tests.
def process(count):
    # BUG: 0 is a valid count, but `if count:` treats 0 as "missing"!
    if count:
        return f"processing {count} items"
    return "skipped (truthiness thinks 0 means 'nothing')"

def process_fixed(count):
    # FIX: ask the precise question — was a real value passed?
    if count is not None:
        return f"processing {count} items"
    return "skipped (truly no value)"

print("buggy  count=0 :", process(0))          # wrongly skipped!
print("fixed  count=0 :", process_fixed(0))    # correctly processed
print("fixed  count=None:", process_fixed(None))

# 🧠 Mental model: `if x:` answers "is x truthy?" (non-empty, non-zero).
#    `if x is not None:` answers "does x exist at all?". Pick the one you mean.
````

### 🎁 Bonus — short-circuit: `and` / `or` return an *operand*, not just `True`/`False`

One more truthiness consequence worth knowing. `and` and `or` don't merely return a boolean — they return **one of the actual operands**, and they stop as soon as the answer is known (*short-circuiting*):

- `a or b` → returns `a` if `a` is truthy, otherwise `b`. (Stops at `a` if it can.)
- `a and b` → returns `a` if `a` is falsy, otherwise `b`.

This powers the classic "default value" idiom `name = user_name or "Anonymous"` — if `user_name` is empty (falsy), you fall back to the default.


````python
# `or` / `and` hand back an operand (evaluated left-to-right, stopping early).
print("'' or 'Anonymous'      →", repr("" or "Anonymous"))     # '' is falsy → 'Anonymous'
print("'Ada' or 'Anonymous'   →", repr("Ada" or "Anonymous"))  # 'Ada' truthy → 'Ada'
print("0 or 42                 →", repr(0 or 42))               # 0 falsy → 42
print("[] and 'never reached'  →", repr([] and "never reached"))  # [] falsy → [] (stops)
print("[1] and 'reached'       →", repr([1] and "reached"))    # [1] truthy → 'reached'

# The default-value idiom in action:
user_name = ""                       # imagine this came from a form the user left blank
display   = user_name or "Anonymous"
print("\ndisplay name →", display)

# ⚠️ Same trap as before: `qty or 10` turns a legitimate 0 into 10. Use it only
# when falsy genuinely means "absent" for your data.
````

## 3. `for` loops — repeat the same operation over a batch

A `for` loop iterates over a *sequence*. Two patterns dominate:

1. **`range(...)`** — when you need to repeat *n* times or count.
2. **Iterate over a list / tuple / string** — when you have actual data.

In AI/automation work you'll spend most of your time in pattern 2: looping over rows, tickets, prompts, batches, or files.

````python
# Pattern 1: range(stop)  — counts 0, 1, 2, ..., stop-1
print("range(5):")
for i in range(5):
    print(f"  i = {i}")

print("\nrange(1, 6):  (start inclusive, stop exclusive)")
for i in range(1, 6):
    print(f"  i = {i}")

print("\nrange(0, 11, 2):  (step)")
for i in range(0, 11, 2):
    print(f"  i = {i}")
````

````python
# Pattern 2: iterate over a batch of work items
prompts = [
    "Summarise this invoice",
    "Classify this email",
    "Extract dates from this contract",
    "Draft a polite refusal",
]
for p in prompts:
    print(f"  → sending prompt: {p!r}")
````

### 🔬 What actually happens when you write `for x in seq:`

A `for` loop feels primitive, but it's **syntactic sugar** — shorthand that Python expands into a smaller, more fundamental mechanism built on two functions:

1. **`iter(seq)`** — asks the sequence for an **iterator**: a little cursor that knows its position.
2. **`next(iterator)`** — pulls the **next** item out, one at a time. When there are no items left, it raises a special signal, `StopIteration`.

So `for p in prompts:` is really this conversation, repeated:

```text
   for p in prompts:
      │
      ▼
   it = iter(prompts)        ← Step 1: get a cursor (once, at the start)
      │
      ▼
   ┌──────────────────────────────────────────────┐
   │  p = next(it)   ← Step 2: pull the next item │ ◀──┐
   │     │                                        │    │
   │     ├─ got an item?  → run the loop body     │    │ repeat
   │     └─ StopIteration? → leave the loop ──────┘  forever…
   └──────────────────────────────────────────────┘        …until StopIteration
```

The loop variable `p` is just "whatever `next()` handed back this time round". When `next()` runs out and raises `StopIteration`, the `for` loop quietly catches it and stops. You never *see* `StopIteration` — the `for` statement hides it for you.


````python
# PROOF: drive the iterator BY HAND — exactly what `for` does for you automatically.
prompts = ["classify", "extract", "summarise"]

it = iter(prompts)          # Step 1: get the iterator (the cursor)
print("type of it:", type(it).__name__)   # list_iterator — NOT a list!

print(next(it))             # Step 2: pull items one at a time
print(next(it))
print(next(it))

# Now the iterator is exhausted. The next pull raises StopIteration —
# the very signal a `for` loop catches to know it's done:
try:
    next(it)
except StopIteration:
    print("StopIteration! — this is the signal `for` uses to stop the loop")
````

### The desugaring — `for` written out as `while True` + `next`

Putting it together: **anything you write as a `for` loop, Python could run as the `while` loop on the right.** They are mechanically identical — the left is just the readable version.

```text
   THE SUGAR (what you write)          THE DESUGARED FORM (what Python runs)
   ────────────────────────────       ────────────────────────────────────────
   for p in prompts:                   it = iter(prompts)
       print(p)                        while True:
                                           try:
                                               p = next(it)        # pull next
                                           except StopIteration:
                                               break               # ran out → stop
                                           print(p)                # the loop body
```

The cell below runs **both** forms and shows their output is character-for-character identical — proof that one is just sugar for the other.


````python
prompts = ["classify", "extract", "summarise"]

print("── normal for loop ──")
for p in prompts:
    print(" ", p)

print("\n── the SAME loop, desugared with iter()/next() ──")
it = iter(prompts)
while True:
    try:
        p = next(it)                 # exactly what `for` calls each round
    except StopIteration:
        break                        # exactly how `for` decides to stop
    print(" ", p)

# 🧠 Mental model: `for x in seq` ≡  "it = iter(seq); keep calling next(it)
#    and running the body until StopIteration says stop." That's the WHOLE loop.
````

### Why this matters — it explains things that otherwise look like magic

Once you see `for` as `iter()` + `next()`, several Python facts stop being mysterious:

- **Why you can loop over strings, dicts, files, and `range(...)` with the *same* `for`** — they all know how to hand back an iterator via `iter()`. The `for` loop doesn't care *what* it's looping over, only that it's *iterable*.
- **Why a generator or a file can be looped once and then is "empty"** — an iterator is a one-way cursor; once `next()` has walked off the end, it stays exhausted. (A *list* isn't exhausted, because `for` calls `iter()` afresh each time, handing you a brand-new cursor.)
- **Why `next()` lets you peek at the first item before looping** — handy when you want to treat a header row differently from the rest.

The cell below shows the same `for` machinery working over four totally different kinds of object, plus the one-shot nature of an iterator.


````python
# ONE mechanism, MANY iterables — because each provides iter()/next():
print("over a string:", [ch for ch in "abc"])           # characters
print("over a range :", [n for n in range(3)])          # 0,1,2 (range is lazy!)
print("over a dict  :", [k for k in {"x": 1, "y": 2}])  # iterates the KEYS

# An iterator is a ONE-WAY cursor: exhaust it once, and it's spent.
it = iter([10, 20])
print("\nfirst  next():", next(it))   # 10
print("second next():", next(it))     # 20
exhausted = list(it)                   # nothing left to collect
print("what's left   :", exhausted)    # []  ← already walked off the end

# But re-iterating the original LIST is fine — `for` calls iter() again,
# giving a fresh cursor every time:
nums = [10, 20]
print("pass 1:", [n for n in nums])
print("pass 2:", [n for n in nums])    # works again — list isn't consumed
````

### `enumerate` — when you also need the index

If you need both the position *and* the value, `enumerate` is your friend. This shows up constantly in batch processing where you want to label each item.

### Accumulating in a loop — the universal data pattern

Sums, counts, min, max, mean — they all share the same recipe: start a running value *outside* the loop, then update it *inside*.

````python
# Daily latencies (seconds) for an LLM call — a week's worth
latencies = [1.84, 1.95, 2.10, 1.78, 2.32, 1.69, 2.04]

total = 0.0
n     = 0
peak  = 0.0

for lat in latencies:
    total += lat
    n     += 1
    if lat > peak:
        peak = lat

mean = total / n
print(f"Calls           : {n}")
print(f"Mean latency    : {mean:.3f}s")
print(f"Peak latency    : {peak:.3f}s")

# Python's built-ins handle this too — useful as a sanity check
print(f"\nDouble-check    : mean={sum(latencies)/len(latencies):.3f}s, peak={max(latencies)}s")
````

> 🎯 **Mental model.** A `for` loop over a sequence is a conveyor belt: each item walks past, you do something with it, and it disappears. You can take notes (accumulators) as items pass.

## 4. `while` loops — repeat *until* a condition becomes false

Use `while` when you don't know in advance how many iterations you need — only when to stop. The classic AI/automation use case is a **retry loop** around a flaky network call.

````python
# Simulating a "call the API, retry on failure" loop
# (we don't actually call an API here — we just demo the pattern)
import random
random.seed(7)   # makes the demo reproducible

max_attempts = 5
attempt      = 0
success      = False

while attempt < max_attempts and not success:
    attempt += 1
    # Pretend the API succeeds 40% of the time
    success = random.random() < 0.4
    print(f"  attempt {attempt}: {'✅ ok' if success else '❌ fail'}")

print()
if success:
    print(f"Succeeded after {attempt} attempt(s).")
else:
    print(f"Gave up after {attempt} attempts.")
````

> ⚠️ **Infinite-loop trap.** A `while` loop *only* stops when the condition becomes false. Make sure the variable in the condition is actually being updated. In Colab/Jupyter you can stop a runaway cell with the ⏹️ button.

> 💡 **Pattern.** Two stop conditions joined by `and` — "keep trying *while* I haven't succeeded **and** I haven't burned through my budget" — is the safest shape for retries.

## 5. `break` and `continue` — fine-grained control

- **`break`** stops the loop entirely (exit early).
- **`continue`** skips the rest of *this* iteration and starts the next.

These let you avoid deeply nested `if`s and keep code readable.

````python
# break: stop searching once you find a match
print("Finding the first low-confidence ticket:")
scores = [0.97, 0.94, 0.91, 0.42, 0.88]
for i, s in enumerate(scores):
    if s < 0.5:
        print(f"  Found at position {i}: {s}")
        break
````

````python
# continue: skip noisy / malformed records
raw_durations = [4.2, -1.0, 7.1, 0.0, 5.8, -3.0, 6.5]   # negatives & zeros are bad data

clean_total = 0
n_valid     = 0
for d in raw_durations:
    if d <= 0:                # treat as invalid
        continue
    clean_total += d
    n_valid     += 1

print(f"Valid samples : {n_valid}")
print(f"Clean mean    : {clean_total / n_valid:.2f}s")
````

## 7. Handling errors gracefully — `try / except`

Real data is messy. Real APIs go down. Instead of crashing the whole pipeline, you can **catch** the error and continue with the next item:

````python
# Cleaning a column of "stringly typed" latencies from a log file
raw = ["1.84", "missing", "n/a", "2.10", "", "1.78", "2.32", "abc", "1.69"]

valid_latencies = []
for s in raw:
    try:
        v = float(s)                # may raise ValueError
        if v <= 0:
            raise ValueError("non-positive")
        valid_latencies.append(v)
        print(f"  OK   {s!r:>10}  →  {v:.2f}s")
    except ValueError as e:
        print(f"  SKIP {s!r:>10}  →  {e}")

print(f"\nKept {len(valid_latencies)} / {len(raw)} valid values")
````

> 💡 Catch **specific** exceptions, not a bare `except:` — bare excepts hide bugs you would actually want to know about. Common ones in data work: `ValueError`, `KeyError`, `TypeError`, `ZeroDivisionError`, `FileNotFoundError`, `requests.RequestException`.

## 8. Putting it all together — a tiny triage pipeline

Let's combine *everything* into one realistic mini-pipeline: route a small batch of incoming customer messages based on (a) a model confidence score, (b) whether the message contains escalation keywords, and (c) whether the customer is a VIP.

Notice how the same three patterns recur:

1. **a loop** over the batch,
2. **branching logic** for the rule,
3. **a `try / except`** to keep one malformed row from killing the run.

````python
# Each ticket: (message, confidence_score, is_vip)
batch = [
    ("How do I export my data?",         0.94, False),
    ("ESCALATE: my account is locked!",  0.31, True),
    ("Can I get a refund?",              0.68, False),
    ("URGENT! billing wrong",            0.55, True),
    ("Hello",                            0.99, False),
    (None,                               0.40, False),     # a bad row
]

ESCALATION_WORDS = ("urgent", "escalate", "lawyer", "refund")

# Counters for the report
counts = {"auto-respond": 0, "draft-for-agent": 0, "human-only": 0, "errors": 0}

for msg, conf, is_vip in batch:
    try:
        # Step 1: a malformed message should be skipped
        text = msg.lower()              # raises AttributeError if msg is None

        # Step 2: classify by rule
        if any(w in text for w in ESCALATION_WORDS) or (is_vip and conf < 0.8):
            decision = "human-only"
        elif conf >= 0.9:
            decision = "auto-respond"
        else:
            decision = "draft-for-agent"

        counts[decision] += 1
        print(f"  {decision:<16}  conf={conf:.2f}  vip={is_vip}  {msg!r}")
    except AttributeError:
        counts["errors"] += 1
        print(f"  SKIP malformed row    msg={msg!r}")

# Step 3: a small summary
print("\nTriage summary:")
for bucket, n in counts.items():
    print(f"  {bucket:<16}: {n}")
````

**What this little script demonstrates.**

- **`try/except`** to survive a malformed input (`None.lower()` would crash without it).
- **`any(... for ...)`** — a *generator expression* — to test "does any keyword match?" in one line.
- **a dictionary used as a counter** (we'll learn dictionaries properly in Notebook 4; for now think of it as a labelled bucket of counts).
- **rule precedence** — escalations and VIPs jump the queue *before* confidence is checked.

This is the entire skeleton of a production triage script. The only difference between this 25-line notebook cell and the production version is that the real one reads from a queue, talks to a model, and writes back to a database. The control flow is the same.

attempt        = 0
max_attempts   = 5

```python
def parse_safely(values):
    out = []
    for v in values:
        if v is True or v is False:          # booleans first — they're also ints!
            out.append(float(v))
        elif isinstance(v, (int, float)):
            out.append(float(v))
        elif isinstance(v, str):
            try:
                out.append(float(v.strip()))
            except ValueError:
                pass                          # silently skip malformed strings
    return out


data = [10, 3.5, "42", "3.7", "n/a", None, "", True, "abc", "  19 "]
cleaned = parse_safely(data)
print(f"Input  : {data}")
print(f"Cleaned: {cleaned}")
```

**Reasoning.** Three habits:
1. Booleans are a subclass of `int` in Python — handle them *first* or
   you'll silently coerce `True` to `1.0` even when you'd rather skip.
2. Strip whitespace before converting.
3. Catch `ValueError` for non-numeric strings — *never* a bare `except:`.

## 🧠 Key takeaways

1. **Indentation defines blocks** — 4 spaces, never mix tabs.
2. `if / elif / else` runs **the first true branch**, in order. Order matters — most restrictive condition first.
3. `for` iterates over a sequence; use `range()` for counters, `enumerate()` when you need indices.
4. `while` repeats until a condition becomes false — make sure that variable changes inside the loop, and protect retries with a max-attempts guard.
5. `break` exits a loop entirely; `continue` skips to the next iteration.
6. **Accumulator pattern**: initialise outside the loop, update inside. Works for sum, count, min, max, peak, …
7. Wrap risky operations in `try / except` and catch *specific* exception types.
8. The skeleton of every automation pipeline is: **loop → branch → handle errors → record outcome**.
