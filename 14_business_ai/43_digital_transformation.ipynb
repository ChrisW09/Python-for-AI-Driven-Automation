## 1. Why AI feels different *this* time

AI has gone through several waves — expert systems in the 1980s, ML in the 2000s, deep learning around 2012, and the current generative-AI wave from ~2020. Three things make the current wave qualitatively different from earlier ones:

1. **The marginal cost of intelligence dropped sharply.** A useful LLM call costs cents, takes seconds, and needs no in-house ML team. Previous waves required either a data-science group or a six-figure consultancy engagement before anything ran.
2. **The interface is natural language.** That's not a technical change — it's an organisational one. Anyone on staff can prototype an AI workflow if they can describe it in writing. The bottleneck moves from *can engineering build it* to *does the business know what to ask for.*
3. **The same model handles many tasks.** Where earlier models were single-purpose (a churn predictor, a fraud detector), today's foundation models cover classification, summarisation, translation, code-generation, and structured extraction with the same API. The economics of "shall we build that?" change accordingly.

> 🧠 **Mental model.** Earlier AI waves added *specific* capabilities at high fixed cost. The current wave adds *general* capability at very low marginal cost. The strategic implication is that the right question shifted from *which AI project should we fund?* to *which of our existing processes should we redesign around the assumption that intelligence is a cheap utility?*


## 2. The unit of change — tasks, not jobs

Meridian's leadership opened a strategy offsite with the wrong question: *"how many support roles can AI replace?"* That framing produces fear and bad plans. So before anything else, let's fix the unit of analysis.

A surprisingly persistent misconception is that AI "replaces jobs." The more accurate (and more useful) framing is that AI **replaces tasks within jobs**. A typical white-collar role is a bundle of 20–40 distinct tasks; AI tends to absorb a handful of them and leave the others untouched.

```
  Job: "Customer support agent"   (45 distinct tasks)

                                       ▼ tasks AI absorbs first
  ┌─────────────────────────────────────────────────────────────┐
  │  Drafting first-response email reply                         │   ◄ low-stakes,
  │  Classifying incoming tickets by topic & urgency             │     repetitive
  │  Summarising long conversation history before escalation     │     pattern-matching
  │  Looking up policy text from the help-centre knowledge base  │
  │  Drafting refund-letter wording from policy + transaction    │
  └─────────────────────────────────────────────────────────────┘
  ┌─────────────────────────────────────────────────────────────┐
  │  Talking down an angry customer on a video call              │   ◄ high-stakes,
  │  Negotiating an exception that requires manager approval     │     judgement,
  │  Spotting fraud that doesn't match any rule yet              │     long-tail
  │  Coaching a junior colleague on tone                         │
  └─────────────────────────────────────────────────────────────┘
                                       ▲ tasks AI is BAD at
```

The first list is automated, the second is amplified. The job continues to exist — but it has fewer routine tasks and more high-judgement ones. That's the right unit for planning a real AI initiative: enumerate the tasks of the affected role, sort them by *how patterned* and *how stake-heavy* they are, and start at the top-left.

> 🏢 **Meridian's turn.** When Meridian's support lead enumerated her agents' day, "support agent" dissolved into ~45 tasks. The top of the first box — *classifying and routing incoming tickets* — was where the queue was clogging and where the time was being lost. That single task, not "the support job," became the candidate for Meridian's first AI initiative. Hold that thought: it's the exact project NB 44 will *structure*, NB 45 will *build*, and NB 46 will *ship* as Case 1.

> 💡 **Practical implication.** When a stakeholder asks *"will AI replace these 12 people?"*, the right answer is usually *"AI will absorb 30–40% of their current task time. The question is whether you reshape the role to absorb the freed-up time productively or treat it as headcount savings — that's a management choice, not a technology one."*

## 3. The five-stage AI maturity model

Knowing *what* to automate isn't enough — you also need to know what your organisation can realistically *absorb* right now. A stage-1 company that tries to ship a stage-3 platform fails not on the AI but on the operations around it. Use this model when you're trying to figure out where an organisation actually is and what the realistic next step looks like. Most organisations sit between stages 2 and 3 in 2026.

| Stage | Name | Signature behaviour | Typical sign you're here |
|---|---|---|---|
| 0 | **Unaware** | No leadership attention; AI not in the strategy doc. | Individual employees use ChatGPT on their phone for work. |
| 1 | **Experimenting** | A few pilots; "innovation lab" / sandbox; results stay in slides. | You can name three pilots but none has reached production users. |
| 2 | **Operationalising** | One or two AI features live in production, owned by a real team. | A daily-active count exists; a roadmap exists; on-call exists. |
| 3 | **Industrialising** | An AI platform team; reusable RAG / model-serving infra; many features. | A new AI feature ships in weeks, not months. |
| 4 | **AI-native** | AI is in the *process design*, not bolted onto it; org chart reflects this. | Job postings list "AI fluency" rather than "experience with X tool." |

**Reading the table.** The hop from stage 0→1 is mostly *permission* (leadership says yes). The hop from 1→2 is the hardest in practice — it requires solving the *boring* problems: monitoring, evaluation, data pipelines, security review, on-call. Most pilots die there. Stage 3 is when the *time-to-second-feature* shrinks because the infrastructure already exists.

> 🏢 **Where is Meridian?** Honestly, **stage 1 leaning into 2.** A couple of teams have toyed with ChatGPT; nothing is in production with an owner and a dashboard. That diagnosis matters: it tells Meridian the realistic next step is *one* production feature done properly (the ticket-triage idea from §2) — not an AI platform, not ten pilots. The whole rest of the module is Meridian making the 1→2 hop, which is exactly the hop where most companies stall.

## 4. Three change strategies — substitution, augmentation, reinvention

You've picked a task and gauged your maturity. The next decision quietly determines how much resistance, risk, and redesign you're signing up for: *how aggressively* does the AI change the existing process? Every AI initiative implicitly picks one of three change strategies. Being explicit about which one you're picking saves a lot of wasted effort later.

```
    SUBSTITUTION                AUGMENTATION                REINVENTION
    ─────────────               ────────────                ───────────
    Same process,                Same process,               Different process,
    AI replaces a step.          AI assists at a step.       redesigned around AI.

    Old: human triages.          Old: human triages.         Old: agent reads ticket.
    New: AI triages.             New: AI suggests,           New: customer's question
         Human reviews edge       human confirms.                 is answered by RAG
         cases.                                                    bot. Ticket never
                                                                  reaches a human
                                                                  unless the bot is
                                                                  uncertain.

    Easy to measure ROI.         Modest ROI, low risk.       Hard to measure ROI,
    Job impact visible.          Job impact hidden.          high upside.
    Resistance common.           Adoption usually smooth.    Requires process,
                                                              org-chart, KPI redesign.
```

> 🎯 **Intuition.** Most real-world AI projects are **augmentation** at year one, drift to **substitution** for the well-understood tasks by year two, and only the most ambitious initiatives reach **reinvention** — and even then only for one or two flagship processes. Telling your stakeholders this up front saves the *"why isn't AI fixing everything yet?"* conversation at the wrong moment.

> 🏢 **Meridian picks one.** Given its stage-1 maturity, Meridian deliberately starts with **augmentation**: the AI *suggests* a ticket category and route, a human still owns the queue. That's the lowest-risk way to build the monitoring-and-evaluation muscle a stage-2 company needs. By NB 46 you'll see Meridian's triage quietly *drift to substitution* for the high-confidence cases once trust is earned — exactly the year-one-to-year-two arc described above.

## 5. The four most common adoption pitfalls

Pitfall 1 — **Technological solutionism.** Buying / building AI for its own sake because everyone else is, without a clear problem statement. The tell: the project pitch starts with "AI" instead of with "users currently waste two hours a week on X."

Pitfall 2 — **Pilot-purgatory.** Pilots ship, look good in slides, are never integrated into a real workflow, and quietly die when the next quarter's priorities arrive. The tell: success metrics are *demo metrics* (accuracy on a curated set) rather than *deployment metrics* (daily active users, time saved, NPS delta).

Pitfall 3 — **Underinvesting in evaluation and monitoring.** The pilot showed 92% accuracy; six months later nobody knows whether it's still 92% because nobody set up the dashboards. AI systems degrade silently when input distributions shift, when the underlying model gets updated by the provider, or when users adapt around the system's quirks. NB 26 was the dedicated technical notebook on this — Module 14 is where you ensure the *organisation* has someone who owns the dashboard.

Pitfall 4 — **Skipping change management.** Treating the AI rollout as a *deployment* problem (we ship, users adapt) instead of a *change* problem (we ship, users need training, incentives, escape hatches, and a fair chance to give feedback that actually changes the product). The tell: post-launch user-survey results were not part of the launch plan.

> ⚠️ **Pitfall #5 — bonus, deserves its own line.** *Treating AI ethics / governance / safety as a separate workstream that runs after the technical project.* It needs to start at the same time as the technical work, with the same stakeholders. Module 14 NB 46 covers this end-to-end.


## 6. 🧠 Mental model — the strategy map

When a stakeholder asks *"where should we use AI?"*, the answer is rarely "build a thing." It's usually a 2x2 grid:

```
                  LOW stakes / errors are cheap         HIGH stakes / errors are costly
                  ────────────────────────              ────────────────────────────
                                                      
  PATTERNED   ► ┌──────────────────────┐    ┌─────────────────────────────────┐
  / repetitive │  YES — automate now.  │    │  YES — but with human in the    │
  task         │  Augment first, then  │    │  loop. Show every confidence,   │
               │  substitute.          │    │  log everything, easy override. │
               └──────────────────────┘    └─────────────────────────────────┘

  ONE-OFF      ► ┌──────────────────────┐    ┌─────────────────────────────────┐
  / judgement   │  Use AI as a draft    │    │  NO. Don't replace the human    │
  task          │  assistant. Save the   │    │  here. Use AI only for         │
                │  human time on framing.│    │  evidence-gathering, never for │
                │                        │    │  the decision itself.          │
                └──────────────────────┘    └─────────────────────────────────┘
```

Quadrant by quadrant: top-left is where most success stories live (NB 21–26 patterns), top-right is most legal / regulated automation (banking decisioning, medical pre-screening), bottom-left is where the *productivity multiplier* for knowledge workers comes from, bottom-right is the legal / ethical land mine you avoid.


## 7. The human side — what changes for the people in the room

Three patterns that show up in every successful AI rollout:

**(a) The job description quietly shifts.** Calling something a *"customer-support agent"* a year after a triage AI lands hides the fact that the role is now *"escalations and edge cases."* Refresh job descriptions, KPIs, and hiring criteria within six months of deployment — otherwise you're hiring the wrong people for what the work has become.

**(b) Skill mix changes.** The *technical* skill demanded of knowledge workers drops at the bottom (the AI does the boring bits) and rises at the top (you need to understand the AI's failure modes well enough to catch them). Plan training accordingly.

**(c) The relationship with the system changes.** Users will *anthropomorphise* the AI — trust it too much when it's confident, ignore it when it's been wrong recently. This is human, predictable, and addressable through interface design (showing confidence, showing evidence, making override frictionless).

> 💡 **Tip.** The single most cost-effective intervention is *publishing the AI's confidence score next to every output it produces.* Users learn to recalibrate within weeks; without it they oscillate between blind trust and total distrust for months.


## 8. 🧠 Mini-recap

- The current AI wave is different because intelligence is cheap, the interface is natural language, and one model serves many tasks.
- The right *unit of change* is the **task**, not the job. Enumerate, score, start at the high-pattern / low-stakes corner.
- The five-stage maturity model (Unaware → AI-native) tells you the *next* realistic step, not the destination.
- Three change strategies — substitution, augmentation, reinvention — have different ROI, risk, and change-management profiles. Pick one explicitly.
- Four pitfalls kill most initiatives: solutionism, pilot-purgatory, no monitoring, no change management.
- People shifts (job description, skill mix, anthropomorphisation) start the day you ship.

### Exercise 1 — ⭐ Spot the *task*, not the *job*

A regional logistics company has 12 dispatchers who plan delivery routes from 06:00 to 14:00 each day. Their work involves: (a) reading the morning customer list, (b) checking traffic / weather, (c) drafting a route plan, (d) handling phone calls from drivers about exceptions, (e) reconciling end-of-day delivery confirmations, (f) writing a daily summary email to operations.

**Identify two tasks AI should absorb first, two tasks you would leave to humans, and explain the criterion you used.**

<details>
<summary>💡 <b>Solution / Answer</b></summary>

**Absorb first:** (a) reading the customer list (basic extraction + classification) and (f) writing the daily summary email (structured generation from logs). Both are highly patterned, low-stakes, and the human cost is hours per week per dispatcher.

**Leave to humans:** (d) handling phone calls about exceptions and (c) drafting the route plan — at least until the AI has been observed for months. Calls require judgement on tone and trust; route planning involves multi-objective tradeoffs (cost, time, driver preferences, customer expectations) that are easy for the AI to get *almost* right and hard for it to get *robustly* right.

**Criterion:** pattern density × stake. High pattern + low stake = absorb first. Low pattern × high stake = leave alone for now.
</details>

A 200-person SaaS company has: a dedicated AI engineering team (3 people), an LLM-powered RAG search bar inside their product (live, ~12,000 queries/day), an internal Slack bot for HR FAQs (live, used daily by ~30 employees), and three more AI experiments in early prototype stage. The CTO says AI is a top-3 strategic priority.

**Place this organisation on the 0–4 maturity scale and justify in two or three sentences. What's the realistic next stage and what would have to happen to get there?**

<details>
<summary>💡 <b>Solution / Answer</b></summary>

**Stage 2 (Operationalising), with one foot in Stage 3.** Two AI features are demonstrably in production with real users (Stage 2 signature behaviour). ### Exercise 3 — ⭐⭐ Pick a change strategy

An insurance company wants to use AI in claims processing. Three options on the table:

- **A:** Replace the first-pass claims agents with an AI triage that auto-approves clear-cut claims and routes the rest to humans.
- **B:** Give every agent an AI co-pilot that drafts the response email and suggests the next action; the agent reviews and clicks.
- **C:** Redesign the claims process around customers self-submitting via a chatbot that handles everything except disputed cases.

**Map each option to substitution / augmentation / reinvention. Pick one for an organisation at Stage 1 (one or two pilots) and justify your choice.**

<details>
<summary>💡 <b>Solution / Answer</b></summary>

**A = substitution.** AI replaces a step (first-pass triage) in the existing process. Humans still do the rest.

**B = augmentation.** AI assists the human at the same step they're already doing. No headcount change implied.

**C = reinvention.** The customer-side process is redesigned; the org chart changes; KPIs change.

**Pick B for a Stage 1 organisation.** Stage 1 has no production AI experience yet; B has the lowest deployment risk (humans are still the decision-makers), and shipping it builds the monitoring / evaluation muscle the organisation needs before tackling A or C. A is the natural next step after 6–12 months of operational data from B. C is a multi-year reinvention that requires a Stage 3+ organisation.
</details>

### Exercise 4 — ⭐⭐ Diagnose the pitfall

Read this short scenario and identify which of the four common pitfalls (solutionism / pilot-purgatory / underinvestment in eval / skipped change management) is most clearly playing out. Defend the call.

> *"We built an AI-powered meeting-summary tool at our 200-person company. We demoed it at the all-hands; the CEO loved it; it went live in the company workspace. Six weeks later, only 14 people are using it weekly. The product manager says the AI's summaries are 'mostly fine.' Engineering is now exploring whether to integrate it with the calendar."*

<details>
<summary>💡 <b>Solution / Answer</b></summary>

**Pilot-purgatory, with overtones of skipped change management.**

Signs of pilot-purgatory: the success metric was the demo ("CEO loved it"), not deployment (only 14 weekly active users — small fraction of a 200-person company). Six weeks of low adoption with no investigation is the signature "quiet death" pattern.

Skipped change management is also in play: nobody trained users on when to invoke it, nobody integrated it into the natural meeting workflow, nobody asked the 186 non-users why. The PM's *"mostly fine"* is the giveaway — they have no idea, because nobody set up evaluation either.

Fix: stop building (calendar integration would compound the problem). Run user interviews with the non-users. Define one daily-active metric. Either pivot the tool to where it's wanted, or sunset it gracefully.
</details>

A junior colleague writes the following claim in a strategy document. Find at least two things that are wrong and write a one-sentence correction for each.

<details>
<summary>💡 <b>Solution / Answer</b></summary>

### Stretch exercise B — ⭐⭐⭐ Build a task-inventory table for a role you know

Pick a role (yours, a family member's, a colleague's). List 12–20 distinct tasks the role does in a typical week. For each task, give two scores out of 5:

- **Pattern density** — how *similar* the task is each time it appears (1 = unique each time, 5 = identical pattern every time).
- **Stake** — what's the cost of an error (1 = nobody notices, 5 = serious consequences for customers / company / safety).

Sort by `pattern × (6 − stake)`. The top three rows are your AI-pilot candidates.

<details>
<summary>💡 <b>Solution / Answer</b></summary>

**Example for a marketing analyst:**

| Task | Pattern | Stake | Score |
|---|---|---|---|
| Pulling Monday metrics from 6 dashboards into a slide | 5 | 1 | 25 |
| Drafting weekly newsletter copy | 4 | 2 | 16 |
| Quarterly board-prep deck | 2 | 5 | 2 |
| Tagging campaigns in the CRM | 5 | 1 | 25 |
| Ad-hoc data investigation | 1 | 3 | 3 |
| … | | | |

Top three (score ≥ 16) are pilot candidates; bottom three (score ≤ 4) are the *don't-touch* list. The middle is your roadmap.

The exercise is itself the point: most knowledge workers have no clear inventory of what their week actually consists of, and most AI initiatives start without one. Building it forces honesty about where the genuine repetitive work lives.
</details>

### Stretch exercise C — ⭐⭐⭐ Counter-argument workshop

A skeptical board member responds to your AI pilot proposal with: *"This is just the next hype cycle. We tried chatbots in 2018 and they were terrible. Why is this time different, and what's our exit ramp if you're wrong?"*

Write a 200-word response. You must (a) take the concern seriously, (b) give a concrete, non-promotional answer to *why this time is different*, and (c) propose a real exit ramp.

<details>
<summary>💡 <b>Solution / Answer</b></summary>

**Example response:**

> *"You're right to be wary — most of the 2018 chatbots failed because the underlying tech couldn't handle the variability of real questions, and the integration cost dwarfed the savings. Two things have changed since then. First, the marginal cost of a single question-answer interaction has dropped roughly two orders of magnitude — what cost €0.50 per call in a 2018 NLU stack costs €0.005 with an LLM today. Second, the same model can handle classification, summarisation, and structured extraction, so we don't need three vendors and three integrations.

> Our exit ramp is the design: this is a co-pilot, not a replacement. The AI suggests; a human approves with one click. If the suggestion quality drops, the human still does the right thing — we just lose the productivity gain. We can switch off the AI in one config flag without a re-architecture. The cost commitment is ~€40k for the pilot quarter; the worst case is we learn something useful about our own process and don't ship the production version. That's a reasonable bet to take."*

What to look for in your own answer: concrete numbers (not vague "things have improved"), a real fallback (not "we'll iterate"), and acknowledging the skeptic's experience rather than dismissing it.
</details>

Pretend it's 18 months from now and the AI initiative you sketched in Stretch A has failed. **Write four plausible failure modes**, in priority order of likelihood. For each, propose one *prevention* (something you'd do before launch) and one *detection* (something that would alert you within four weeks of the problem starting).

<details>
<summary>💡 <b>Solution / Answer</b></summary>

**Example pre-mortem** (continuing the support-triage pilot from Stretch A):

| # | Failure mode | Prevention | Detection |
|---|---|---|---|
| 1 | Re-routing rate creeps from 4% to 15% as the input distribution drifts (new product launch changes vocabulary). | Publish a weekly *category coverage* dashboard. Add a process for re-labelling 50 hard cases monthly. | Alert if re-routing rate > 8% in any 7-day window. |

- The human side starts the day you ship: job description, skill mix, anthropomorphisation.
