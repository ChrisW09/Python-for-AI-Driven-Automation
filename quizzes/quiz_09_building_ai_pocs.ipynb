### Q1 — Tokens

NB 27 says an LLM does not see words; it sees **tokens**. The practical consequence most relevant to cost and limits is...

- **A.** Tokens are always whole words
- **B.** Text is split into sub-word tokens, and both context limits and pricing are counted in tokens, not words
- **C.** Tokens only matter for non-English text
- **D.** One token equals one character

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

Tokenization splits text into sub-word units (~4 chars / ~0.75 words on average in English). Context windows *and* API bills are denominated in tokens — so 'how long can the prompt be?' and 'what will this cost?' are both token questions. NB 27 makes you count them rather than guess.
</details>


### Q2 — Next-token prediction

An LLM generates text by...

- **A.** Looking the answer up in a database
- **B.** Repeatedly predicting the most likely next token given everything so far, one token at a time
- **C.** Translating from an internal language
- **D.** Running a search engine query

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

At its core the model is an autoregressive next-token predictor: it produces a probability distribution over the next token, samples one, appends it, and repeats. NB 27 builds intuition for why this makes models fluent yet prone to confident fabrication — there is no lookup, only likelihood.
</details>


### Q3 — POC complexity ladder

NB 34 builds three POCs of growing complexity. The pedagogical point of starting with a **Streamlit-only** app is...

- **A.** Streamlit is the best production framework
- **B.** Prove the idea with the least architecture first; add a backend/DB/model only when a real need appears
- **C.** You should never use a backend
- **D.** Streamlit is the fastest at runtime

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

The ladder (Streamlit → +FastAPI+SQLite → +ML model → §7 two-service split) embodies 'smallest architecture that proves the idea'. Each step is justified by a need the previous step exposed. Jumping straight to microservices for a prototype is exactly the over-engineering NB 50's decision table warns against.
</details>


### Q4 — RAG

In the RAG pipeline of NB 35, why retrieve context and put it in the prompt instead of relying on the model's own knowledge?

- **A.** It makes the model run faster
- **B.** It grounds answers in your specific, current documents and reduces hallucination — the model cites what you gave it
- **C.** It removes the need for any LLM
- **D.** It permanently retrains the model on your data

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

RAG injects *relevant, current, private* text at query time so the model answers from your corpus rather than its frozen training data — cutting hallucination and letting you cite sources. It does **not** retrain the model (that's fine-tuning) and still needs the LLM to synthesize the retrieved chunks.
</details>


### Q5 — When not to use an agent

NB 36's honest take on agentic AI is that you should prefer a fixed pipeline over an autonomous agent when...

- **A.** The task is open-ended with unpredictable steps
- **B.** The steps are known and fixed — then a plain script is cheaper, faster and more reliable than letting an agent plan
- **C.** You have a large budget
- **D.** You want the system to look impressive

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

NB 36 asks 'does the agent earn its keep?' An agent's value is *deciding which tools to call in what order* under uncertainty. If the workflow is deterministic, that planning overhead buys nothing and adds failure modes — a straight-line script wins. Agents pay off for genuinely branching, unpredictable tasks.
</details>
