### Q1 — Safer HTTP

Which `requests.get(...)` call is **most robust** in production code?

- **A.** `requests.get(url)`
- **B.** `requests.get(url, timeout=5)`
- **C.** `requests.get(url, timeout=5).json()`
- **D.** `requests.get(url, timeout=5, headers={'User-Agent': 'me/1.0'})`

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: D.**

Three production-grade habits at once: an explicit `timeout` (so a stalling server doesn't hang your job forever), a sensible `User-Agent` (so the API owner can identify you in their logs), and the structured call shape. C is also good but skips the UA.
</details>


### Q2 — SQL parameterisation

Why is `cur.execute("SELECT * FROM users WHERE name = %s", (name,))` better than `cur.execute(f"SELECT * FROM users WHERE name = '{name}'")`?

- **A.** It's faster
- **B.** It avoids SQL injection
- **C.** It returns more rows
- **D.** It works on more databases

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

Parameterised queries pass `name` as **data**, not as part of the SQL string. The string `name = '; DROP TABLE users; --` is just a long username, not an executable statement. f-string interpolation into SQL is the classic SQL-injection vulnerability.
</details>


### Q3 — `json()` parsing

An API returned `"OK"` (just the two letters). Will `response.json()` succeed?

- **A.** Yes — strings are valid JSON
- **B.** No — JSON must always be an object `{}`
- **C.** It will return `None`
- **D.** It depends on the Content-Type header

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: A.**

JSON values include string, number, boolean, null, array, and object — not just objects. `"OK"` is a valid JSON string. The misconception that JSON must be `{...}` is widespread but wrong.
</details>


### Q4 — HTTP status

Your code does `response.raise_for_status()`. What does it do?

- **A.** Always raises an exception
- **B.** Raises only on 4xx / 5xx status codes
- **C.** Raises only on connection errors
- **D.** Raises only when the response body is empty

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

`raise_for_status()` is a one-line guard: it does nothing on 2xx (success) and 3xx (redirect, but requests follows by default), and raises `HTTPError` on 4xx / 5xx. It's much safer than `if response.status_code == 200:` because it correctly handles 201, 204, etc.
</details>


### Q5 — Schema validation

With Pydantic, you declare `class User(BaseModel): age: int`. What happens if the API returns `{"age": "30"}`?

- **A.** Raises a ValidationError
- **B.** Returns `age=30` (Pydantic coerces the string to int)
- **C.** Returns `age="30"` (the type hint is ignored)
- **D.** Returns `age=None`

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

By default Pydantic v2 coerces compatible types — strings that look like integers become integers. If you want strict typing (and a ValidationError instead of coercion), use `Field(strict=True)` or `BaseModel.model_config = ConfigDict(strict=True)`.
</details>
