### Q1 — Prompt structure

When designing a classification prompt, where should the **system instructions** go?

- **A.** In the user message
- **B.** In the system message
- **C.** In a separate function call
- **D.** After the examples

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

The system role is for stable behaviour ("you are a classifier; output JSON"). The user role is for the task at hand. Modern LLM SDKs all support this split and treat the system message specially — for example, Anthropic puts it in a top-level `system=` parameter, not in the message list.
</details>


### Q2 — Validating LLM output

Your prompt asks the model for JSON. Sometimes it returns extra prose around the JSON. The cheapest robust fix is to...

- **A.** Use a different model
- **B.** Wrap `json.loads` in `try/except` and retry once with a stricter prompt
- **C.** Parse the prose manually with regex
- **D.** Disable temperature

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

Catch the `JSONDecodeError`, ask the model again with `"Reply with strict JSON only."`. Set `temperature=0` while you're at it. For production-grade reliability, also validate the parsed JSON against a Pydantic schema before using it downstream.
</details>


### Q3 — Embedding similarity

Two L2-normalised embeddings `a` and `b`. What does their dot product `a @ b` represent?

- **A.** Euclidean distance
- **B.** Cosine similarity
- **C.** L1 distance
- **D.** Cross-entropy

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

When `||a|| = ||b|| = 1`, the dot product equals the cosine of the angle between them — that's *cosine similarity*. This is why most embedding APIs return L2-normalised vectors: you get cosine-similarity for the price of a single dot product, no division needed.
</details>


### Q4 — RAG basics

What is the **retrieval** step in Retrieval-Augmented Generation actually doing?

- **A.** Calling the LLM with a longer context
- **B.** Fetching the top-k relevant documents from a vector store before calling the LLM
- **C.** Re-training the LLM on your documents
- **D.** Caching previous LLM responses

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

RAG separates retrieval (a fast similarity search over your documents) from generation (the LLM call). The retrieved documents are stuffed into the prompt as context. It's the cheapest way to give an LLM access to private / current data without fine-tuning.
</details>


### Q5 — Agent loop

An LLM-powered agent uses *tools* in a loop. The single most important thing about each tool is that it...

- **A.** Runs fast
- **B.** Has a well-defined input/output schema the LLM can reliably emit and the runner can reliably parse
- **C.** Is written in Python
- **D.** Has access to the internet

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

Agents fail at the **interface** more often than at the **algorithm**. A tool whose schema is ambiguous makes the LLM either skip it or call it wrong. Pydantic models or JSON-Schema definitions per tool are now the default approach (OpenAI's function-calling, Anthropic's tool-use, LangChain's tools all enforce this).
</details>
