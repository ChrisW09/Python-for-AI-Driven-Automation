### Q1 — Train/test split

Why do we split data into train and test sets?

- **A.** To make training faster
- **B.** To estimate how the model will perform on new, unseen data
- **C.** Because scikit-learn requires it
- **D.** To reduce memory usage

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

The test set is held-out data the model has never seen — it's the only honest way to estimate generalisation. Training accuracy is misleading because the model has already seen those examples. Without a test set, you can't tell overfitting from real learning.
</details>


### Q2 — When to scale features

You're using `LogisticRegression` with a `StandardScaler`. Where should you fit the scaler?

- **A.** On the whole dataset before splitting
- **B.** On the training set only, then transform both train and test
- **C.** On the test set
- **D.** On train and test separately

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

Fit on train, transform both. Fitting on the whole dataset leaks information from the test set into the scaler's mean/std (data leakage). Fitting separately on train and test means the test data is transformed with different parameters — that's wrong too. The Pipeline construct enforces the right order automatically.
</details>


### Q3 — Imbalanced classification

On a dataset where 95% of labels are `0`, a model that always predicts `0` will have what accuracy?

- **A.** 50%
- **B.** 0%
- **C.** 95%
- **D.** Undefined

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: C.**

It's 95% — and worthless. This is the classic case where accuracy is a misleading metric. Look at the confusion matrix, precision, recall, F1, ROC-AUC, or PR-AUC instead.
</details>


### Q4 — Threshold tuning

A classifier returns `predict_proba(...)`. To tune the trade-off between false positives and false negatives, you...

- **A.** Retrain with `class_weight="balanced"`
- **B.** Pick a custom probability threshold instead of the default 0.5
- **C.** Use a more complex model
- **D.** Add more features

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

The default 0.5 threshold is a **convention**, not an optimum. When the cost of FP differs from the cost of FN, the optimal threshold shifts. Plot the PR curve and pick the threshold that minimises your expected cost. This is what NB 15 §5 walks through.
</details>


### Q5 — Target leakage

Which of these features will most likely cause **target leakage** in a model predicting whether a customer will churn next month?

- **A.** The customer's signup date
- **B.** The number of support tickets in the last 3 months
- **C.** Whether the customer has cancelled their subscription
- **D.** The customer's plan tier

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: C.**

If the customer has already cancelled, the feature "has cancelled" perfectly predicts "will churn" — because it's the target itself. Target leakage is when a feature directly or indirectly encodes the label. A model with leakage looks brilliant in training and is useless in production.
</details>
