### Q1 — Why AI now

A manager asks why AI automation is worth doing *now* rather than five years ago. The strongest single answer from NB 43 is...

- **A.** Computers finally got fast enough
- **B.** LLMs made many language/knowledge tasks doable without bespoke ML, collapsing the cost of automating them
- **C.** AI is now perfectly reliable
- **D.** Cloud storage became cheap

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

NB 43 frames the shift as *economic*, not magical: general-purpose models turned tasks that used to need a dedicated ML project (and a dataset, and a team) into a prompt away. That collapse in marginal cost — not raw hardware or any claim of perfection — is what changes the build-vs-skip calculation. The notebook is also explicit that reliability is *not* solved (hence Modules 6/21 on evaluation).
</details>


### Q2 — Architecture sizing

NB 44's decision table says you should reach for a **microservices** architecture when...

- **A.** Always — it's the modern default
- **B.** The app has more than one file
- **C.** Parts of the system have genuinely different change cadence, scaling needs, or team ownership
- **D.** You are building a prototype

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: C.**

NB 44 §5 (and the §7 split in NB 28) tie service boundaries to *real differences* — change cadence, scaling profile, dependencies, team ownership. Splitting by nouns without such a difference produces a 'distributed monolith': all the cost of the network, none of the benefit. For a POC, a single 3-tier service is usually the right size.
</details>


### Q3 — Git & pull requests

In the NB 45 collaborative workflow, the point of a **pull request** is to...

- **A.** Back up your code to the cloud
- **B.** Get a second person to review a change before it merges to the main branch
- **C.** Run the code faster
- **D.** Avoid having to write commit messages

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

A PR is a *review gate*: someone other than the author reads the diff before it lands on `main`. NB 45 pairs this with the 60-second review checklist and the four AI-failure modes — the human review is exactly where AI-generated code gets caught. Backup is a side effect of pushing, not the purpose of a PR.
</details>


### Q4 — RACI for AI

In NB 46's RACI for an AI initiative, who is ultimately **Accountable** for a model's decisions in production?

- **A.** The model itself — it made the decision
- **B.** The vendor of the LLM
- **C.** A named human (e.g. the product owner) — accountability never transfers to the AI
- **D.** Nobody, if the model is autonomous

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: C.**

NB 46 §2 is emphatic: AI can be *Responsible* (it does the work) or *Consulted*, but a **human is always Accountable**. 'The model decided' is not a defensible answer in a governance review or an audit. This is the principle the EU AI-Act-style obligations and the NB 18 fairness aside both rest on.
</details>


### Q5 — POC vs MVP vs Production

NB 46 insists POC, MVP and Production are *three projects, not three phases*. The clearest reason is...

- **A.** They use different programming languages
- **B.** Each answers a different question — 'is it possible?', 'is it useful to real users?', 'is it reliable at scale?' — and success at one does not imply the next
- **C.** Each must be built by a different company
- **D.** Production is just a bigger POC

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

A POC proves *feasibility* (often throwaway code); an MVP tests *usefulness* with real users; Production is about *reliability, monitoring and scale*. NB 46 stresses that a dazzling POC routinely fails to become a useful MVP — treating them as one continuous build is how the §4 Case 3 project quietly died.
</details>
