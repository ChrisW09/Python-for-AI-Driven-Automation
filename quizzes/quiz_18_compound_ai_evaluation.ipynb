### Q1 — The attribution problem

A team ships one release that swaps the model, adds retrieval, and rewrites the prompt. The eval score jumps from 0.8 to 2.8. NB 53 calls the claim "the new model did it" a...

- **A.** Valid conclusion — the score went up, the model was upgraded
- **B.** Confounded comparison — several factors changed at once, so their effects are entangled in one before/after number
- **C.** Type II error
- **D.** Pareto improvement

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

Three knobs turned in one release means the single before/after delta contains all three effects (plus their interactions) mixed together — a *confounded comparison*. NB 53 §1's whole point: no amount of staring at the aggregate number separates them; you need a factorial experiment that varies each factor systematically. (In NB 53's own study, the "obvious" hero — the 5× model — turned out to matter far less than retrieval.)
</details>

### Q2 — Why replicate?

NB 53 runs every configuration × question pair **6 times**. The reason is...

- **A.** To make the study look more rigorous
- **B.** Jupyter caches results, so single runs are unreliable
- **C.** LLM outputs are nondeterministic — replication lets statistics separate real factor effects from run-to-run noise
- **D.** The judge needs 6 answers to produce a score

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: C.**

A single run of config A beating config B is one coin flip, not a conclusion — NB 53 §5 shows three *different* configs "winning" depending on which single replication you look at. Replication gives every comparison an error bar, and the ANOVA's F-test then asks whether the between-config differences exceed what noise alone would produce.
</details>

### Q3 — Design size

A pipeline has four binary factors: retrieval ∈ {none, keyword}, reranker ∈ {off, on}, model ∈ {small, large}, prompt ∈ {terse, grounded}. A **full factorial** design runs...

- **A.** 4 configurations — one per factor
- **B.** 8 configurations — you can drop one factor
- **C.** 16 configurations — every combination of levels (2⁴)
- **D.** 256 configurations

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: C.**

Full factorial = the Cartesian product of all levels: 2 × 2 × 2 × 2 = 16. Every added binary factor doubles the grid — which is exactly why NB 53 §8 introduces **fractional factorial** designs (run a structured half or quarter of the grid, keep main effects estimable, sacrifice only high-order interactions) once you're past ~4 factors.
</details>

### Q4 — The right model for the rubric

Scores come from an ordinal 0–3 rubric. NB 53 warns that plain OLS/ANOVA on these scores is an approximation because...

- **A.** OLS only works with negative numbers
- **B.** An ordinal scale doesn't promise equal spacing between levels — the scale-correct model is a cumulative-link (ordinal) mixed model, which CAFE fits via R
- **C.** ANOVA cannot handle more than two factors
- **D.** Rubric scores must always be converted to percentages first

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

"3 = correct, 1 = vague" is an *ordering*, not a number line — nothing guarantees the 2→3 gap equals the 0→1 gap, and OLS assumes it does. For a quick offline read the approximation is fine, but near a decision boundary you want the scale-matched model: **CLMM** for ordinal rubrics, logistic for binary pass/fail — precisely the "scale-matched models" CAFE runs through R's `ordinal`/`lme4` packages.
</details>

### Q5 — Reading the Pareto frontier

On NB 53's quality-vs-cost chart, a configuration sits **off** the Pareto frontier. That means...

- **A.** It is the cheapest option available
- **B.** Its judge score is below 2.0
- **C.** It failed the significance test
- **D.** Some other configuration is both cheaper *and* better — running this config is strictly a mistake

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: D.**

Pareto-optimal = not dominated: no other config beats it on *both* axes at once. Everything off the frontier is dominated — you could pay less and get more. In NB 53's study the `none/large` config (expensive model, no retrieval) is the cautionary example: paying 5× for the model without giving it context is dominated by the cheap model *with* retrieval. The business conversation then happens only among frontier points.
</details>
