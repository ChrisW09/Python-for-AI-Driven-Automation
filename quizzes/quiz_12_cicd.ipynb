### Q1 — Docker build cache

You edit only your application code (not `requirements.txt`) and rebuild the image. The Dockerfile is ordered `COPY requirements.txt` → `RUN pip install` → `COPY app`. Which layers rebuild?

- **A.** Every layer, from `FROM` onward
- **B.** Only the `COPY app` layer and everything above it — `pip install` stays cached
- **C.** Only the base image
- **D.** None — Docker never rebuilds an existing image

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

Docker's cache is busted from the **first changed layer upward**. Because code is copied *after* the dependency install, editing code leaves the expensive `RUN pip install` layer cached and only the cheap layers above rebuild. That is exactly why you copy `requirements.txt` first and application code last.
</details>

### Q2 — Compose startup order

In `docker-compose.yml` the backend declares `depends_on: db: condition: service_healthy`. What does that guarantee?

- **A.** The backend starts a fixed number of seconds after the db
- **B.** The backend waits until the db's **healthcheck passes** before it starts
- **C.** The db restarts whenever the backend crashes
- **D.** Nothing — `depends_on` is only documentation

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

Plain `depends_on` only waits for a dependency to **start**; `condition: service_healthy` waits for its healthcheck to **pass**. Without it the backend can launch before Postgres is accepting connections and crash-loop — the classic 'it worked, then the DB wasn't ready' bug.
</details>

### Q3 — Registry tags vs digests

You deploy `myapp:latest` today, and again next month from the same tag. Why might the two deployments run *different* code?

- **A.** They can't — an image tag is immutable
- **B.** `:latest` is a **mutable pointer**; a later `push` moves it. Only a digest (`@sha256:…`) is immutable
- **C.** The registry randomly picks an image
- **D.** Docker deletes old layers after a month

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

A tag such as `:latest` is just a movable label — pushing again repoints it at a new image. For reproducible deploys, pin an **immutable digest** (`myapp@sha256:…`) or at least an immutable version tag like `v1.4.2`. 'Deploy latest' is 'deploy whatever happens to be latest right now.'
</details>

### Q4 — GitHub Actions job graph

A workflow has a `test` job and a `deploy` job, with `deploy:` declaring `needs: test`. What happens on a push?

- **A.** `test` and `deploy` run in parallel
- **B.** `deploy` runs only after `test` succeeds; if `test` fails, `deploy` is skipped
- **C.** `deploy` runs first, then `test`
- **D.** Both run only on a schedule, never on push

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

`needs:` adds a dependency edge to the job **DAG**. `deploy` waits for `test` and runs only if it passed — so a red build never ships. Jobs with no `needs` relationship run concurrently, which is how a matrix of tests fans out.
</details>

### Q5 — The reverse proxy

In the module's example app, only **nginx** publishes a host port (80/443); the backend and database stay on the private `appnet` network. What is nginx doing?

- **A.** Acting as the database
- **B.** Acting as a **reverse proxy** — the single public front door that routes requests to internal services and terminates TLS
- **C.** Acting as the container registry
- **D.** Acting as the DNS server for the domain

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

The reverse proxy is the one public entry point. It terminates **HTTPS/TLS** and forwards each request to the right internal service over the private network. Your application and database are never exposed to the internet directly — a big security win, and the reason DNS points the domain at the proxy, not at the app.
</details>
