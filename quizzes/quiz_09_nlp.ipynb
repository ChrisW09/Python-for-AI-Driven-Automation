### Q1 — The librarian's four moves

NB 35 describes BERTopic's default pipeline as four modular stages. In order, they are...

- **A.** Tokenise → count words → LDA → pick $K$
- **B.** Embed (SentenceTransformer) → reduce (UMAP) → cluster (HDBSCAN) → describe (c-TF-IDF)
- **C.** Cluster (HDBSCAN) → embed → reduce (UMAP) → describe
- **D.** Reduce (UMAP) → describe (c-TF-IDF) → embed → cluster

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

The *automatic librarian*: **read the gist** (a SentenceTransformer turns each doc into a ~384-dim vector where semantic similarity is geometric closeness), **declutter** (UMAP squeezes to ~5 dims, because density clustering fails in high dimensions), **build the shelves** (HDBSCAN finds however many dense blobs exist — no $K$ up front), **label the spines** (c-TF-IDF treats each cluster as one super-document and finds its distinctive words). The stages are independent and swappable — that modularity is BERTopic's real design win.
</details>

### Q2 — 30% of my documents are in topic `-1`!

You fit BERTopic on real support tickets and nearly a third of the documents land in **topic `-1`**. Per NB 35, the right reaction is...

- **A.** Panic — the model failed; refit until `-1` disappears
- **B.** Delete those documents; they must be duplicates
- **C.** Nothing is wrong: `-1` is HDBSCAN's outlier/noise bin — an *honesty feature*; call `reduce_outliers()` afterwards only if you deliberately need every document labelled
- **D.** Treat `-1` as the most important topic, since it is the biggest

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: C.**

Topic `-1` is the librarian's **"misc" cart**: documents that sit in no dense region (off-topic rants, one-offs, gibberish). HDBSCAN refuses to force them onto a shelf — unlike LDA or KMeans, which *must* assign every document somewhere, smearing junk across real topics and diluting them. If you do need every document labelled, `reduce_outliers(docs, topics)` reassigns them to the nearest topic — a deliberate, separate decision rather than a silent default.
</details>

### Q3 — What STREAM actually is

NB 36 introduces **STREAM** (`stream-topic`). Its central design point is...

- **A.** A faster reimplementation of BERTopic
- **B.** A real-time service that streams tweets into LDA
- **C.** A single new neural topic model that beats all classical ones
- **D.** A *unified API* over many topic models — classical (`LDA`, `NMF`), neural (`ProdLDA`, `ETM`, `CTM`) and clustering (`KmeansTM`, `CEDC`) all train via the same `fit` / `get_topics` pattern, so swapping models is a one-line change

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: D.**

STREAM is the *test kitchen*: same ingredients (your corpus in a `TMDataset`), swappable chefs (the models), same stove (`model.fit(dataset, n_topics=...)` then `model.get_topics()`). Trading `KmeansTM()` for `ProdLDA()` changes one constructor line — which makes *honest model comparison* cheap instead of betting the analysis on a single algorithm. Probabilistic models additionally expose `get_beta()` (P(word|topic)) and `get_theta()` (the per-document topic mixture you can feed downstream).
</details>

### Q4 — Judging the dishes

Per NB 36, the honest way to decide whether a topic model is any good is...

- **A.** Eyeball the top-words lists — if they look convincing, ship it
- **B.** Report both the classic word-cooccurrence coherence (`NPMI`) *and* the embedding-based metrics (`ISIM`, `INT`, `ISH`, `Expressivity`), and hand-inspect topics where they disagree
- **C.** Pick the model with the most topics
- **D.** Accuracy on a held-out test set

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

Top-words lists *look* convincing even when a model is bad, so you quantify instead of trusting your eyes. `NPMI` is cheap and language-agnostic but rewards words that merely co-occur (it can be fooled by boilerplate); the embedding metrics ask the harder, more human question — *do these words mean similar things?* — and catch "junk" topics NPMI misses. A senior practitioner reports **both** and treats large disagreements as a signal to inspect the topic by hand. (For choosing the topic count, `optimize_and_fit` sweeps `k` under an information criterion — a starting point to inspect, not gospel.)
</details>

### Q5 — Which rung of the sentiment ladder?

NB 37's ladder: lexicon (VADER) → classical ML (TF-IDF + LogisticRegression) → transformer. You have a few hundred labelled in-domain reviews. The default professional choice is...

- **A.** The transformer — always start with the most powerful model
- **B.** VADER — training a model is never worth it
- **C.** An unevaluated LLM prompt
- **D.** The TF-IDF + LogisticRegression **workhorse** — cheap, fast, interpretable via `.coef_`, and hard to beat in-domain; escalate to a transformer only if it isn't good enough

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: D.**

The golden rule of the ladder: **climb only as high as the problem forces you to.** With labels available, the classical model learns *your* domain's vocabulary (a generic lexicon doesn't know that "outrageous" means "too expensive" in your review pile), trains in seconds, predicts in microseconds on CPU, and `.coef_` shows exactly which words drive each prediction — invaluable for debugging and stakeholder trust. Transformers earn their size and latency only on hard, context-heavy text (negation, subtle phrasing) — and then you pick a *domain-matched* pretrained model (FinBERT for finance, twitter-roberta for social), because domain shift is a classic cause of bad results.
</details>
