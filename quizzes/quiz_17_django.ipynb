### Q1 — The MTV request cycle

A request arrives at a Django app. In what order do the pieces handle it?

- **A.** Template → Model → View
- **B.** **URLconf → View → (Model + Template) → Response**
- **C.** View → URLconf → Model → Template
- **D.** Model → View → URLconf → Template

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

Django matches the URL via the **URLconf** to a **view**; the view pulls data through the **Model** (ORM) and renders a **Template** into the `HttpResponse`. That is the MTV (Model–Template–View) cycle — Django's take on MVC.
</details>

### Q2 — Models & migrations

You add a new field to a Django model. What has to happen before the database actually has that column?

- **A.** Nothing — Django alters the table automatically at runtime
- **B.** Run `makemigrations` to generate a migration, then `migrate` to apply it
- **C.** Hand-edit the SQLite file
- **D.** Drop and recreate the whole database

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

The model *describes* the schema in Python. `makemigrations` turns a model change into a versioned migration file; `migrate` applies pending migrations to the database. Migrations are reviewable in code review and repeatable across dev/staging/prod — that's the whole point over hand-edited SQL.
</details>

### Q3 — Form validation

In a Django form, when is it safe to read `form.cleaned_data`?

- **A.** Any time after you construct the form
- **B.** Only after `form.is_valid()` has returned `True`
- **C.** Before calling `is_valid()`, to pre-fill defaults
- **D.** Only from inside a template

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

`cleaned_data` is populated by validation, so it exists **only after `is_valid()` runs**. Reading it first raises `AttributeError`. The canonical pattern is always: build the form → `is_valid()` → then use `cleaned_data`, or re-render the page with `form.errors`.
</details>

### Q4 — Serving a model

A Django view scores a customer with a trained scikit-learn model. Where should the model be loaded?

- **A.** Freshly from disk inside the view, on every request
- **B.** **Once at startup** (module import / app-ready), then reused for every request
- **C.** Retrained from the data on each request
- **D.** Inside the template that renders the result

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

Load the artifact **once** when the process starts and keep it in memory; each request just calls `.predict`. Loading from disk per request is slow, and *training* per request is the classic blunder — training belongs offline, serving loads the saved artifact.
</details>

### Q5 — Authentication

A prediction-history page must be visible only to logged-in users. The simplest correct way in Django is…

- **A.** Check `request.user` by hand in the view and every template
- **B.** Decorate the view with `@login_required` — anonymous users are redirected to the login page
- **C.** Put the page behind a password typed into the HTML
- **D.** Only link to it from pages you trust

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

`@login_required` is Django's built-in decorator: anonymous requests are redirected to the login page, authenticated ones pass through. Django's auth system (users, sessions, login/logout views, password hashing) does the heavy lifting — you should not roll your own.
</details>
