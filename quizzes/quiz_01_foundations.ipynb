### Q1 — Variable assignment

Which of these is the **correct** way to assign the value `42` to a variable called `n`?

- **A.** `n == 42`
- **B.** `n = 42`
- **C.** `n := 42`
- **D.** `42 = n`

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

`=` is assignment; `==` is comparison. `:=` (the walrus operator) does assignment-as-expression but isn't used in standalone statements like this. `42 = n` is a syntax error — you can't assign *to* a literal.
</details>


### Q2 — f-string format spec

What does the f-string `f"{0.873:.1%}"` produce?

- **A.** `0.873%`
- **B.** `87.3%`
- **C.** `87%`
- **D.** `0.9`

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

`:.1%` multiplies the value by 100, formats it with one decimal, and appends `%`. So `0.873` becomes `87.3%`. Memorise this — every dashboard you'll ever build needs it.
</details>


### Q3 — List behaviour

After `xs = [1, 2, 3]` and `xs.append(4)`, what is the value of `xs`?

- **A.** `[1, 2, 3]` — `append` returns a new list
- **B.** `[1, 2, 3, 4]` — `append` mutates in place
- **C.** `[4, 1, 2, 3]`
- **D.** `None`

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

`list.append` **mutates in place** and returns `None`. This is a common beginner trap: writing `xs = xs.append(4)` will leave `xs == None`. Just call `xs.append(4)`.
</details>


### Q4 — Dict lookup

If `d = {"a": 1}` and you write `d["b"]`, what happens?

- **A.** Returns `None`
- **B.** Returns `0`
- **C.** Raises `KeyError`
- **D.** Adds `"b": None` to the dict

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: C.**

A missing key on `d[key]` raises `KeyError`. To get `None` (or any default), use `d.get("b")` or `d.get("b", 0)`. To set on miss, use `defaultdict` or `dict.setdefault`.
</details>


### Q5 — Function semantics

Inside a function, you assign a new value to a parameter. Does that change the caller's variable?

- **A.** Always
- **B.** Never
- **C.** Only for mutable types (list, dict)
- **D.** Only if the parameter has a default value

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

Reassigning a *parameter* never changes the caller's variable — the parameter is just a local name. But if you **mutate** a mutable object the parameter points to (`xs.append(...)` on a passed list), the caller does see that — because they're holding a reference to the same object.
</details>
