### Q1 — Package layout

Which of these is the **recommended modern Python project layout**?

- **A.** A single `.py` file at the project root
- **B.** A `src/<package>/` directory with `pyproject.toml` at the root
- **C.** Notebooks committed alongside scripts
- **D.** A `lib/` folder full of free-standing modules

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

The `src/` layout prevents Python from accidentally importing your in-progress code from the repo root. `pyproject.toml` is the modern packaging spec (PEP 621) — it replaces `setup.py` and `setup.cfg`. Together they make `pip install -e .` work reliably and CI happy.
</details>


### Q2 — Retries

Your scheduled job calls a flaky API. The right retry policy is...

- **A.** Try once, log on failure
- **B.** Try 100 times in a tight loop
- **C.** Try a few times with exponential backoff and jitter, capped
- **D.** Try forever

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: C.**

Exponential backoff (`1s, 2s, 4s, 8s, …`) avoids hammering an already-struggling service. **Jitter** (a random multiplier) prevents many retriers from synchronising on the same retry boundary — the classic thundering-herd problem. A retry cap prevents runaway loops in production. `tenacity` is the standard library to do all this in a decorator.
</details>


### Q3 — Secrets

You need an OpenAI API key in your scheduled script. Where should it live?

- **A.** Inside the .py file as a string
- **B.** Inside a `config.yaml` you commit to git
- **C.** In an environment variable read by the script at startup
- **D.** In a comment near the top of the file

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: C.**

Secrets in environment variables (or a secrets manager in real deployments), never in source. Use `python-dotenv` for local dev convenience — but `.env` belongs in `.gitignore`. Committed YAML configs are fine for **non-secrets** (defaults, host names).
</details>


### Q4 — Cron expression

What does `0 6 * * 1-5` mean?

- **A.** Every minute on weekdays
- **B.** 6 AM every weekday (Monday-Friday)
- **C.** 6 PM on the 1st-5th of the month
- **D.** Every hour for the first 6 days of the month

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

Fields are: minute hour day-of-month month day-of-week. So `0 6 * * 1-5` = at minute 0 of hour 6, any day, any month, day-of-week 1 through 5 (Monday–Friday). Quick check: `crontab.guru` reads them in plain English for you.
</details>


### Q5 — Logging vs print

In a production script that runs nightly, what should report progress?

- **A.** `print(...)`
- **B.** Comments
- **C.** The `logging` module with structured levels
- **D.** Writing to a temp file

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: C.**

`logging` gives you levels (DEBUG / INFO / WARNING / ERROR), timestamps, structured output, and per-module loggers — all of which your monitoring system (or future you debugging at 3 AM) will need. `print(...)` is fine in notebooks; in scheduled jobs it goes to stdout undated and you'll regret it.
</details>
