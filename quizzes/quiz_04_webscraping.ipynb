### Q1 — Scrape, or don't?

You need data from a website. Before you write a single BeautifulSoup selector, what is the right first move — and what must you check before you fetch anything?

- **A.** Just start parsing the rendered HTML — an API is always slower and more restrictive than reading the page yourself.
- **B.** Check for an official API, data export, or bulk dataset first; only if none exists do you scrape — and then you read `robots.txt` and the Terms of Service *before* you fetch.
- **C.** Read `robots.txt` only after you get blocked — it's advisory and only matters once a server returns `403`.
- **D.** Skip the API question and go straight to a managed scraper like Firecrawl, since it handles every site.

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

Scraping is a **last resort**, not a first move: walk the list and stop at the first "yes" — an official API or export (OpenAlex is the model case), then a bulk dataset / RSS / sitemap, and *only then* the rendered HTML. An API is stabler, faster, and sanctioned. And once you *do* scrape you're a guest on someone else's server, so you check `robots.txt` and the ToS **before** you knock — not after you're blocked.
</details>

### Q2 — `find` returned nothing

A book card has no discount badge, yet your extractor runs `book.find("span", class_="discount").text`. What happens, and why?

- **A.** It returns an empty string `""` — `.text` on a missing tag is just blank.
- **B.** It returns `None` silently, so the row gets a `None` discount and nothing breaks.
- **C.** It raises `AttributeError: 'NoneType' object has no attribute 'text'` — `find` returned `None`, and `None` has no `.text`.
- **D.** It raises a `KeyError`, because `class_="discount"` isn't a valid attribute filter.

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: C.**

`find` returns the **first** matching node, or **`None`** when there's no match (whereas `find_all` and `.select()` return a possibly-empty *list*). Chaining `.text` onto that `None` raises `AttributeError: 'NoneType' object has no attribute 'text'` — *the* classic scraping bug, and exactly how a silent layout change surfaces at runtime. Guard it: grab the node first, then read its text only `if node else default`.
</details>

### Q3 — Being a polite guest

Which set of habits makes a scraper a **polite guest** rather than an abusive bot?

- **A.** Fire as many parallel requests as possible to finish fast, and retry hard the instant you see a `429`.
- **B.** Rotate through random User-Agents and proxies so you never trip the rate limit.
- **C.** Spoof a real browser's User-Agent, ignore `robots.txt` (it's unenforceable anyway), and skip caching so the data is always fresh.
- **D.** Send an honest `User-Agent` with a contact URL, throttle between requests (honouring any `Crawl-delay`), cache responses so re-runs don't re-fetch, and obey `robots.txt`.

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: D.**

A well-behaved scraper does four things every time: **identify** itself with an honest `User-Agent` (a contact URL lets an admin email you instead of just banning you), **throttle** with `time.sleep` and honour any `Crawl-delay`, **cache** every response so a re-run never re-fetches, and **obey `robots.txt`** before it knocks. Hammering with unbounded parallel hits or dodging limits with rotating proxies is precisely the behaviour that gets your IP blocked — a `429 Too Many Requests` is the server asking you to slow down, not to try harder.
</details>

### Q4 — Firecrawl or do-it-yourself?

You're choosing between `requests` + BeautifulSoup and a managed scraper like Firecrawl. When does the managed API clearly earn its keep — and when is it overkill?

- **A.** Always reach for Firecrawl — hand-written scrapers are obsolete and can't produce clean data.
- **B.** Use Firecrawl for a JavaScript-rendered or anti-bot page, or when you want schema-driven **typed JSON**; for a simple static page, three lines of `requests` + BeautifulSoup still win.
- **C.** Use Firecrawl only for static HTML, and switch to `requests` + BeautifulSoup the moment a page needs JavaScript.
- **D.** They're interchangeable — the choice is purely about which library you happen to have installed.

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

A managed scraper earns its keep exactly where DIY breaks down: **JavaScript-rendered** pages (`requests` only gets the empty shell), **anti-bot** walls, crawling at scale, or when you want **schema-validated typed JSON** with no selectors to maintain. For a static, well-structured page, `requests` + BeautifulSoup in a few lines is free and wins — don't reach for the heavy, *billed* tool when three lines would do. Option **C** has the escalation backwards: static → DIY, dynamic / anti-bot → managed.
</details>

### Q5 — Paging through OpenAlex

You need a large result set out of OpenAlex — well beyond a single page. What's the correct, scalable pagination pattern?

- **A.** Increment `page=1,2,3…` until you've got everything; it scales to any number of results.
- **B.** You can't page OpenAlex without an API key — request one first and pass it as the `cursor`.
- **C.** Start with `cursor=*`, then on each response follow `meta['next_cursor']`, stopping when it comes back `None` (with a loop cap as a seatbelt).
- **D.** Ask for everything at once with `per-page=100000`; OpenAlex has no page-size limit.

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: C.**

For large exports OpenAlex uses **cursor** paging: send `cursor=*` on the first request, then pass back the `meta['next_cursor']` value each time until it returns `None` — always with a loop cap so a forgotten cursor-advance can't hammer the server. Plain `page=` only reaches the first 10,000 results and a single response caps at `per-page=200`, so neither **A** nor **D** scales. And no key is required at all — you just add your email as `mailto=` to join the faster **polite pool** — so **B** is wrong too.
</details>
