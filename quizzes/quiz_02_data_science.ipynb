### Q1 — Groupby semantics

What does `df.groupby("region")["revenue"].sum()` return?

- **A.** A Series indexed by region, values are summed revenues
- **B.** A DataFrame with columns `region` and `revenue`
- **C.** A list of revenue sums
- **D.** A single number

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: A.**

`groupby(key)[col].agg(...)` returns a Series whose index is the group key and whose values are the aggregated quantity. To get a DataFrame instead, append `.reset_index()` or use double brackets: `df.groupby("region")[["revenue"]].sum()`.
</details>


### Q2 — Missing data

Which of these reliably removes rows where `df["score"]` is missing?

- **A.** `df[df["score"] != None]`
- **B.** `df[df["score"].notna()]`
- **C.** `df[df["score"] != np.nan]`
- **D.** `df[df["score"] != ""]`

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

Comparing to `np.nan` is **always False** (`np.nan != np.nan`), so option C silently removes nothing. `None` comparisons are similarly broken on float columns. Always use `.notna()` / `.isna()` / `pd.notna(x)`.
</details>


### Q3 — Broadcasting in NumPy

If `a` has shape `(3, 1)` and `b` has shape `(1, 4)`, what is the shape of `a + b`?

- **A.** `(3,)`
- **B.** `(4,)`
- **C.** `(3, 4)`
- **D.** Error — shapes don't match

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: C.**

NumPy broadcasting expands a `(3, 1)` array and a `(1, 4)` array to a common `(3, 4)` shape — the `1` dimensions are stretched to match the other operand. This is how outer-product-style computations get written in one line.
</details>


### Q4 — Plotting in a few lines

`monthly` is a DataFrame whose index is the month and whose four columns are support channels. What is the **quickest** way to draw one line per channel, with a legend?

- **A.** A `for`-loop over the columns with `ax.plot(...)`, then `ax.legend(...)`
- **B.** `monthly.plot(title="Automation rate by channel")`
- **C.** `plt.lines(monthly)`
- **D.** `sns.heatmap(monthly)`

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

A DataFrame knows how to draw itself: `.plot()` uses the index as the x-axis, draws every column as its own line, and builds the legend from the column names — one line of code. The matplotlib loop (A) produces the same chart but in ~10 lines of boilerplate; reserve raw matplotlib for plain arrays and final tweaks (the `Axes` that pandas/seaborn return). C doesn't exist, and a heatmap (D) is for matrices, not trends.
</details>

### Q5 — Confidence intervals

You compute a 95% confidence interval for the mean and get `[1.2, 1.8]`. What does this mean?

- **A.** 95% of values are between 1.2 and 1.8
- **B.** There is a 95% probability the true mean is between 1.2 and 1.8
- **C.** If we re-sampled many times, 95% of the resulting CIs would contain the true mean
- **D.** 95% of our sample is between 1.2 and 1.8

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: C.**

This is the textbook subtle distinction. A 95% CI is a statement about the **procedure**: of all the CIs you could compute from independent samples, 95% of them would contain the true mean. The naïve "there's a 95% chance" reading is Bayesian — strictly speaking, frequentist CIs are not probability statements about the parameter.
</details>
