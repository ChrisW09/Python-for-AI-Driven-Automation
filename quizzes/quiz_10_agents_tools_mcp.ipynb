### Q1 — Why every agent loop needs a budget

NB 37 insists every agent loop sets `max_steps`. Without it, the most likely failure is...

- **A.** The agent answers too quickly
- **B.** The model refuses to call tools
- **C.** A loop that calls the same tool forever — burning time and tokens with no stop condition
- **D.** The tools return the wrong type

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: C.**

An agent decides its *own* next step, so nothing inherently stops it — a bad decision (or a tool that never satisfies the goal) loops indefinitely. `max_steps` (and ideally a separate `max_tool_calls`/token cap) is the guardrail. NB 37's rule: *budget every loop, validate every input, log every step, critique the output.*
</details>

### Q2 — ReAct vs plan-and-execute

NB 37 contrasts **ReAct** with **plan-and-execute**. The key advantage of planning *first* is...

- **A.** It never calls tools
- **B.** The agent's intent is inspectable (and approvable) *before* any tool runs
- **C.** It is always faster than ReAct
- **D.** It removes the need for a step budget

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

Plan-and-execute writes an ordered list of sub-goals up front, so you can read, validate, or require approval of the plan before spending a single tool call. ReAct decides one step at a time — more flexible for open-ended tasks, but it can wander. Neither removes the need for a budget.
</details>

### Q3 — The most important field on a tool

In NB 38, when an agent has twenty tools, what actually lets the model pick the right one?

- **A.** The order the tools were registered
- **B.** The tool's `description` — it must say *when* to use the tool
- **C.** The function's source code
- **D.** The tool's return type

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

The model only sees each tool's **name, description, and JSON-Schema** — not your code. The *description* is what it matches against the user's intent, so "a tool is only discoverable if its description says when to use it." NB 38 also stresses validating arguments and returning a consistent `{ok, result|error}` envelope rather than leaking a raw traceback.
</details>

### Q4 — MCP primitives and who controls them

NB 39 says an MCP server exposes **tools**, **resources**, and **prompts**. Who controls each, respectively?

- **A.** The model, the application, the user
- **B.** The user, the model, the application
- **C.** All three are controlled by the model
- **D.** The server controls all three

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: A.**

**Tools** are *model-controlled* (the model decides to call them — like a POST). **Resources** are *application-controlled* (the host reads them by URI — like a GET/file). **Prompts** are *user-controlled* (reusable templates the user invokes — like a slash command). They all travel over **JSON-RPC 2.0** with the lifecycle `initialize` → `*/list` → `*/call|read|get`.
</details>

### Q5 — When to reach for multiple agents

Per NB 40, you should split a task across multiple agents when...

- **A.** Always — more agents are always better
- **B.** Never — one agent can do anything
- **C.** Sub-tasks need different tools/instructions, benefit from independent review, run in parallel, or map to different owners
- **D.** Only when the dataset is large

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: C.**

Multi-agent is a cost (more calls, latency, failure modes), so it only pays when sub-tasks genuinely differ — different tools/prompts, a writer + independent critic, parallelisable work, or distinct permissions. NB 40's rule of thumb: *start with one agent; split only when a single prompt is doing two jobs badly.*
</details>
