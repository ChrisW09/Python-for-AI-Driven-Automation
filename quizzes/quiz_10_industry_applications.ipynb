### Q1 — Probability is not a decision

NB 35 turns a churn probability into a *targeting decision* using a per-customer break-even threshold. Why is a single global 0.5 cutoff wrong?

- **A.** 0.5 is too high for everyone
- **B.** The break-even probability depends on each customer's CLV — high-value customers are worth targeting at much lower risk
- **C.** Probabilities are never trustworthy
- **D.** You should target every customer

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

The expected-value rule targets customer *i* when p·save_rate·CLV > offer_cost, i.e. above p* = offer_cost / (save_rate·CLV). Because CLV varies, p* varies per customer — an Enterprise account is worth a call at 4% risk while a Basic account is not. `.predict()`'s silent 0.5 is the wrong threshold for *every* customer.
</details>


### Q2 — The accuracy trap

A fraud model is reported as 99.6% accurate (NB 36). The right reaction is...

- **A.** Ship it — that's excellent
- **B.** Be suspicious: at ~0.5% fraud, predicting 'not fraud' for everything already scores ~99.5%; ask for PR-AUC and precision/recall@k
- **C.** Retrain with more layers
- **D.** Accuracy is the only metric that matters

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

With extreme class imbalance, accuracy is dominated by the majority class — doing nothing scores ~99.5%. NB 36 reports PR-AUC for model comparison and precision@k / recall@k / €-prevented for operations, because those reflect catching the rare positives, which is the entire point.
</details>


### Q3 — Scale before k-means

NB 37 insists you **scale features before k-means**. Skipping scaling on raw RFM columns produces...

- **A.** Faster clustering
- **B.** Clusters dominated by whichever feature has the largest numeric range — usually just bands of revenue
- **C.** More interpretable clusters
- **D.** No change at all

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

K-means uses squared Euclidean distance, so a feature with a 100× larger range (monetary, in euros) drowns out recency and frequency — the algorithm effectively sorts by revenue. Standardizing gives every axis equal say. The NB 37 Debug-me proves it: raw clusters map almost one-to-one onto revenue quartiles.
</details>


### Q4 — Order the right quantity

NB 38's newsvendor rule orders the demand quantile at the **critical ratio** Cu/(Cu+Co). If under-stocking costs €4/unit and over-stocking €1/unit, you should order...

- **A.** The mean forecast
- **B.** The 80th percentile of demand (ratio = 4/5)
- **C.** The 20th percentile
- **D.** Exactly last week's demand

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

Critical ratio = Cu/(Cu+Co) = 4/(4+1) = 0.8, so you order the 80th percentile, not the mean. Expensive shortages push the order *above* the mean; expensive leftovers (fresh food) pull it below. NB 38 shows by simulation that ordering the mean is beaten by the critical-ratio quantile whenever Cu ≠ Co.
</details>


### Q5 — Respect time

Across NB 35/36/38 the recurring evaluation discipline is...

- **A.** Always use a random train/test split
- **B.** Respect time — temporal splits and shifted rolling features — or your offline metrics overstate what production will deliver
- **C.** Use as many features as possible
- **D.** Optimise accuracy above all

<details>
<summary>💡 <b>Answer + reasoning</b></summary>

**Correct answer: B.**

Churn, fraud and maintenance all predict the *future*, so the honest test is a temporal split, and rolling features must be shifted so they only use information available before the prediction moment. NB 38's Debug-me (unshifted rolling mean) and NB 36's random-split consultant both show how ignoring time manufactures a score that evaporates in production.
</details>
