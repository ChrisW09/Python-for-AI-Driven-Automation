## 3. How each notebook is structured

Every main numbered notebook follows the same six-section template:

| Section | What's in it | Purpose |
|---|---|---|
| 🎯 Learning objectives | 4-6 bullets at the top of the notebook | Tells you what you'll be able to *do* by the end |
| 📚 Numbered concept sections | Short prose → runnable code | The actual teaching |
| 🧪 **Practice exercises** (1, 2, 3, …) | 3–5 per notebook | Apply the concept *while it's fresh* |
| 🐞 Debug-me cell | Usually exercise 4 | A cell that **intentionally errors** — fix it |
| 🧠 **Stretch exercises** (A, B, C, D) | 4 per notebook | Deeper / interview-grade questions |
| 🎁 Bonus mini-project | One per notebook | A larger applied task that ties the chapter together |
| ✅ Self-assessment + 🚀 Next | The closing checklist | Know when to move on |

**Two-tier exercise design.** Every notebook has **two layers** of exercise:

- **Practice** (numbered 1, 2, 3, …) — testing the concept just taught. Short and direct.
- **Stretch** (lettered A, B, C, D) — deliberately harder. The kind of question you'd expect in a take-home, an interview, or a small piece of real work. New (C, D) were added in the 2026 refinement pass — they push you further than the original two stretches did.

**Every single exercise** has a `<details>`-collapsed worked solution with a *Reasoning* paragraph that explains **why** the solution looks the way it does — not just the code.

Across the 22 main numbered notebooks: ~94 practice + ~88 stretch = **180+ exercises** with worked solutions.

## 8. Tips for actually finishing the course

Most self-paced courses fail not because they're hard but because they're hard to *sustain*. Some habits that have helped real learners get to the end:

1. **Set a small, fixed cadence.** 30–45 minutes most days beats 4 hours on Saturday. Notebooks are designed so each *section* is a natural stopping point — you don't need to finish a whole notebook in one session.
2. **Do the exercises *before* opening the solution.** Five minutes of genuine struggle beats five hours of passive reading. The `<details>` block hides the solution for exactly this reason.
3. **Treat the 🐞 Debug-me cell as a puzzle, not a bug.** Every Debug-me cell carries a banner — *'this cell intentionally errors'* — to remind you.
4. **Predict the output before re-running.** Change one number, then *predict* what the new output will be before you hit Shift-Enter. Wrong predictions are where real understanding lives.
5. **Track your progress in this notebook.** Re-run the estimator in Section 7 each weekend with your actual completed notebooks — watching the remaining-time number shrink is surprisingly motivating.
6. **Don't skip the bonus mini-project.** The bonus is where the chapter's ideas leave the playground and meet a realistic scenario. It's also the easiest hour of the notebook to scroll past — which is exactly why it's worth doing.
7. **Resist the appendix rabbit hole until you finish the backbone.** The 11 A-* appendices are tempting but optional; come back to them after you've shipped the capstones.
