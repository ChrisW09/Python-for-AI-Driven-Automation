## 3. GitHub Copilot — the four modes

Copilot exposes four interaction modes; knowing which to use is half the skill.

| Mode | What it does | Best for |
|---|---|---|
| **Inline completion** | Ghost text suggestion as you type. *Tab* to accept, *Esc* to reject. | Boilerplate, loops, repetitive patterns. |
| **Copilot Ask** | Q&A side panel — "explain this function", "why does this error?". | Understanding code, debugging questions. |
| **Copilot Plan** | Generates a multi-step plan, then implements it. | Medium-sized tasks with a clear sub-structure. |
| **Copilot Agent Mode** | Autonomously: creates files, runs terminal commands, fixes errors. | End-to-end POC builds — *our default for the rest of Module 9*. |

### Agent Mode vs Ask Mode at a glance

| Property | Ask Mode | Agent Mode |
|---|---|---|
| Suggest code | ✓ | ✓ |
| Answer questions | ✓ | ✓ |
| Create / edit files | ✗ | ✓ |
| Run terminal commands | ✗ | ✓ |
| Multi-step tasks | ✗ | ✓ |
| Self-correct errors | ✗ | ✓ |
| Whole-project context | limited | ✓ |

> 🧠 **Rule of thumb:** *Ask* for questions, *Agent* for tasks.


### Activating Agent Mode in VS Code

1. Open Copilot Chat (`Ctrl+Shift+I` or the side panel icon).
2. In the dropdown at the bottom of the chat, switch from **Ask** to **Agent**.
3. Type your prompt. The agent can now create files, run commands, and iterate.

**What Agent Mode actually does for you:**

- Creates, edits, and deletes files — across multiple files at once.
- Runs terminal commands (`pip install`, `streamlit run`, `git`, etc.).
- Searches the codebase to build context from the whole project.
- Plans multi-step tasks and organises sub-steps itself.
- Notices its own errors (e.g. import not found) and fixes them.


## 4. 🧠 The vibe-coding loop

The defining shift from classical coding to AI-assisted coding is *what gets iterated*.

```
  CLASSICAL                              AI-ASSISTED (vibe coding)
  ─────────                              ─────────────────────────
  read docs → learn syntax → write       describe goal → AI generates
  code → debug → Stack Overflow →        code → review & run → describe
  re-write...                            the adjustment → repeat

  Each loop: minutes to hours.          Each loop: seconds to minutes.
```

The vibe-coding cycle in four steps, run as fast as you can think:

```
      ┌──────────────────┐    ┌──────────────────┐
      │ 1. Describe idea │ ─► │ 2. AI generates  │
      │    in NL         │    │    code          │
      └──────────────────┘    └──────────────────┘
             ▲                          │
             │                          ▼
      ┌──────────────────┐    ┌──────────────────┐
      │ 4. Describe the  │ ◄── │ 3. Review & run  │
      │    adjustment    │    │                  │
      └──────────────────┘    └──────────────────┘
```

**The trap to avoid:** treating Copilot as a fancy autocomplete. The big productivity gain comes from running the *whole loop*, including the *describe the adjustment* step — many learners stop after step 3 and end up with a brittle blob of generated code they don't fully understand.

> 🎯 **Speed isn't the only point.** Faster loops mean you can *try more variants* before committing. The cost of trying a third architecture sketch went from "half a day" to "thirty seconds". Architecture decisions get better when exploration is cheap.

## 5. Anatomy of a good code prompt — the 5 building blocks

The single biggest predictor of generated-code quality is *prompt specificity*. A vague prompt produces generic code; a specific prompt produces *your* code.

**Every Agent-Mode prompt should include all 5:**

1. **Stack** — name the libraries and frameworks explicitly. *"FastAPI + SQLAlchemy + SQLite"* not *"a web API"*.
2. **Files & folders** — give the layout. *"`backend/main.py`, `backend/models.py`, `frontend/app.py`"* not *"a Python project"*.
3. **Data model & endpoints** — concrete schemas. *"Table `uploads` with columns `id, filename, n_rows, uploaded_at`. Endpoint `POST /api/upload`."*
4. **Demo data path** — *"include a `sample_data.py` that seeds the DB with 500 synthetic rows so the app works on first run."*
5. **Constraints** — CORS, ports, allowed libraries, error behaviour. *"CORS allow `http://localhost:8501`. Pydantic schemas. Standard library only — no external dependencies beyond requirements.txt."*

> 💡 **Rule of thumb:** *bad prompts produce generic code; good prompts produce *your* code.*


### Bad prompt vs good prompt — side by side

**Bad** (vague, produces generic SQLite + Flask sample):

```
Build me an app that can analyse CSVs and do a bit of ML.
```

**Good** (all 5 building blocks, produces *your* app):

```
Create backend/ with FastAPI + SQLAlchemy + SQLite. Models: Upload, AnalysisRun,
ColumnStat. Endpoints: POST /api/upload, GET /api/uploads, POST /api/analyze/{id}.
CORS for http://localhost:8501. Pydantic schemas. Demo data via sample_data.py
that seeds 500 synthetic rows on first start.
```

The second prompt names the stack (FastAPI/SQLAlchemy/SQLite), the data model (three tables), the endpoints (three REST routes), conventions (CORS, Pydantic), and a demo path (sample_data.py). The agent has almost nothing left to guess — and the output reliably matches what you wanted.


## 6. The Git-first workflow — GitHub before code

Most tutorials show you Git *after* you've written code (`git init` locally, then push to remote). For POCs, the reverse order is faster and cleaner: **create the GitHub repo first, then clone, then code.**

```
  1. GitHub repo    →  2. Clone     →  3. Open in     →  4. Create files →  5. Commit & push
     (online)            (git clone)    VS Code           in VS Code        (regularly)
```

**Why this order:**

- README, `.gitignore`, and license are generated by GitHub directly — no boilerplate to remember.
- The remote is connected from second one — no follow-up `git remote add origin ...`.
- Issues, branches, and (later) collaborators are ready from minute one.

### What this looks like in practice

```bash
# After creating the repo on github.com:
git clone https://github.com/<user>/<repo>.git
cd <repo>

# Create a fresh virtual environment
python -m venv .venv
source .venv/bin/activate     # macOS/Linux
# .venv\Scripts\activate     # Windows

# ... write code, run, iterate ...

# Commit regularly
git add .
git commit -m "feat: initial Streamlit scaffold"
git push
```

> 💡 **The golden rule:** commit after *every working change*. Five small commits beat one giant one — easier to review, easier to revert, easier to explain.


### Repo structure for a multi-POC course

If you're doing all three POCs from NB 34 in one repo, the recommended layout:

```
poc-repo/
├── .gitignore             # what NEVER gets versioned
├── README.md              # overview of all three POCs
├── poc1_streamlit_basic/
│   ├── .venv/             # local virtual environment (gitignored)
│   ├── app.py             # Streamlit app
│   ├── requirements.txt
│   └── data/              # sample CSVs
├── poc2_streamlit_fastapi_sqlite/
│   ├── .venv/
│   ├── frontend/          # Streamlit UI
│   ├── backend/           # FastAPI + SQLite
│   ├── requirements.txt
│   └── app.db             # SQLite file (gitignored)
└── poc3_xgboost_churn/    # similar layout plus model.pkl
```

**Conventions:**

- **One repo, three subfolders, one per POC** — easy to clone, one issue board.
- **Each POC has its own `.venv` and `requirements.txt`** — no dependency conflicts between POCs.
- **A shared `README.md` at the root** — short description, setup steps, and run command per POC.

> 🎯 **What goes in the repo:** *only* your own code and what someone else needs to reproduce it on a fresh machine — source code, `requirements.txt`, config templates, docs. **Not** `.venv/`, `__pycache__/`, generated `*.db` or `*.pkl`, or any `.env`.


## 7. `.gitignore` and secrets — getting this wrong is expensive

A `.gitignore` file at the repo root tells Git which files to skip. **Without one, your repo bloats, slows, and leaks secrets.**

**Minimal `.gitignore` for a Python POC:**

```
.venv/
__pycache__/
*.db
*.pkl
.env
*.key
.DS_Store
```

**Categories:**

- **Environment / machine-specific:** `.venv/`, `__pycache__/`, `.DS_Store`, `*.pyc`.
- **Generated at runtime:** `*.db`, `*.pkl`, `models/`, anything the code recreates from sources.
- **Secrets:** `.env`, `.env.*`, `*.key`, `*.pem`, anything an API key could live in.

### Secrets discipline

> ⚠️ **Never commit secrets.** API keys, tokens, passwords, real `.env` files, database files with personal data.

**If it happens anyway:**

1. **Rotate the key immediately.** Deleting from Git history is not enough — it's already on someone's clone or in GitHub's mirrors.
2. Notify your security / DPO team if the key gave access to anything material.

**Prevention:** verify `.gitignore` *before your first commit*. The first commit is the one that sets the pattern — get it right.


## 8. ⚠️ The top 6 vibe-coding pitfalls

Every learner hits some of these in their first few projects. Knowing them in advance saves hours.

1. **Secrets in the repo** — `.env` with a real API key gets pushed. *Mitigation: `.gitignore` before first commit; rotate immediately if you slipped.*
2. **Missing `.gitignore`** — `.venv/`, `__pycache__/`, `*.db` get committed. Repo bloats, clones are slow, generated files cause merge conflicts. *Mitigation: copy the standard `.gitignore` template (§7) into every repo.*
3. **Wrong Python environment** — Conda vs. venv vs. system Python; the agent installs into one and you run from another. Errors say *"module not found"* even though you definitely installed it. *Mitigation: always activate the same venv before installing and before running; verify with `which python`.*
4. **Port conflicts** — Streamlit's 8501 or FastAPI's 8000 already taken. Server starts but you see a stale page. *Mitigation: read the terminal output every time; if conflicts persist, change the port with `--port 8502`.*
5. **CORS errors** — Streamlit (8501) calls FastAPI (8000), browser blocks the cross-origin request. *Mitigation: add `CORSMiddleware` allow `http://localhost:8501` on the backend (see NB 34 POC 2 prompt).*
6. **Huge undocumented commits** — *"updates"* as message, 200 files changed. Nothing reviewable, nothing revertable. *Mitigation: commit after every working change; the commit message reads as a sentence ("Add retry logic to the LLM client").*

> 💡 **Golden rule:** *Setup first, code second.* Half an hour of fastidious setup saves half a day of mysterious errors.

### Exercise 5 — ⭐⭐ Debug me 🐞 (review-style)

A colleague is mid-vibe-coding session and complains: *"I asked Copilot to add caching to my function. It did, the app runs, but the cache hits aren't working — every call re-computes."* They show you the agent's diff:

```python
def expensive_fetch(user_id):
    @lru_cache(maxsize=128)
    def inner(uid):
        return _api_call(uid)
    return inner(user_id)
```

Find at least two things that are wrong with the agent's solution.

<details>
<summary>💡 <b>Solution / Answer</b></summary>

**Issues:**

1. **`@lru_cache` is defined *inside* the function.** Every call to `expensive_fetch` creates a *new* `inner` function with a *new* cache. The cache never persists across calls, so nothing is ever a "hit". *Fix:* move the decorator and the cached function out to module level, or just decorate `expensive_fetch` itself.

2. **Missing `from functools import lru_cache`.** The code would fail at `NameError` on the first call — the agent hallucinated that the symbol was already imported (Failure mode A from NB 30: hallucinated APIs).

3. *(Bonus)* **No `typed=` consideration.** If `user_id` could be passed as int and str interchangeably, `lru_cache` treats them as different cache keys. For an HTTP-style API this can cause confusing cache misses. *Fix:* normalise to a single type at the function boundary.

**The wider lesson:** Copilot output *looks plausible* — it includes the right decorator, the right argument. The bugs are in the *placement* of the decorator and a missing import. This is exactly why the *review* step of the vibe-coding loop is non-negotiable.
</details>
