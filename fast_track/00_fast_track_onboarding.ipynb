## Environment check

Run the cell below. It will confirm your Python version and import the libraries the fast track needs. If anything errors, you can install the dependencies in one line:

```
!pip install numpy pandas matplotlib scikit-learn
```

````python
import sys, importlib, platform

# A tiny helper that imports one package and reports the result.
# (Functions are taught in NB 4 — for now, just read the body line by line.)
def check(pkg):
    try:
        m = importlib.import_module(pkg)
        return f"  ✓ {pkg:<14} {getattr(m, '__version__', '?')}"
    except ImportError:
        return f"  ✗ {pkg:<14} MISSING — install with `!pip install {pkg}`"

print(f"Python    : {sys.version.split()[0]} ({platform.platform()})")
print(f"Executable: {sys.executable}")
print()

# Check each library by name, one explicit line per library.
print(check("numpy"))
print(check("pandas"))
print(check("matplotlib"))
print(check("sklearn"))

print("\n✅ Fast-track environment ready.")
````

## How to use these notebooks well

1. **Run every code cell as you read.** Don't skim — touching the keyboard matters.
2. **Do the Practice exercises before opening the solution.** Five minutes of struggle is worth five hours of passive reading.
3. **The 🐞 Debug-me cell intentionally errors.** It's not a bug; it's a puzzle. Read the traceback, fix the cell, learn the habit.
4. **Predict the output before re-running.** Change one number, predict, then check. Wrong predictions are where real learning happens.
5. **Don't skip ahead too far.** The notebooks build on each other; NB 6 (pandas) assumes NB 4 (functions).
