## 3. GitHub Copilot — the four modes

Your setup is done; now meet your crew. An architect doesn't hand every job to the same worker — you call a different specialist for "fill in this gap" than for "build me the whole east wing". Copilot's four modes are exactly that: four ways of working, each suited to a different size of task. Knowing which to reach for is half the skill.

| Mode | What it does | Best for |
|---|---|---|
| **Inline completion** | Ghost text suggestion as you type. *Tab* to accept, *Esc* to reject. | Boilerplate, loops, repetitive patterns. |
| **Copilot Ask** | Q&A side panel — "explain this function", "why does this error?". | Understanding code, debugging questions. |
| **Copilot Plan** | Generates a multi-step plan, then implements it. | Medium-sized tasks with a clear sub-structure. |
| **Copilot Agent Mode** | Autonomously: creates files, runs terminal commands, fixes errors. | End-to-end POC builds — *our default for the rest of Module 7*. |

### Agent Mode vs Ask Mode at a glance

| Property | Ask Mode | Agent Mode |
|---|---|---|
| Suggest code | ✓ | ✓ |
| Answer questions | ✓ | ✓ |
| Create / edit files | ✗ | ✓ |
| Run terminal commands | ✗ | ✓ |
| Multi-step tasks | ✗ | ✓ |
| Self-correct errors | ✗ | ✓ |
| Whole-project context | limited | ✓ |

> 🧠 **Rule of thumb:** *Ask* for questions, *Agent* for tasks. Building ChurnScope from nothing is a *task*, so Agent Mode is our default — but we'll still pop into *Ask* whenever we want to understand what it produced.

- Notices its own errors (e.g. import not found) and fixes them.

## 4. 🧠 The vibe-coding loop

You've met the crew; now learn the rhythm you'll work in. This is the architect-and-builder model in motion — and the single most important habit in the whole module.

The defining shift from classical coding to AI-assisted coding is *what gets iterated*.

```
  CLASSICAL                              AI-ASSISTED (vibe coding)
  ─────────                              ─────────────────────────
  read docs → learn syntax → write       describe goal → AI generates
  code → debug → Stack Overflow →        code → review & run → describe
  re-write...                            the adjustment → repeat

  Each loop: minutes to hours.          Each loop: seconds to minutes.
```

The vibe-coding cycle in four steps, run as fast as you can think:

```
      ┌──────────────────┐    ┌──────────────────┐
      │ 1. Describe idea │ ─► │ 2. AI generates  │
      │    in NL         │    │    code          │
      └──────────────────┘    └──────────────────┘
             ▲                          │
             │                          ▼
      ┌──────────────────┐    ┌──────────────────┐
      │ 4. Describe the  │ ◄── │ 3. Review & run  │
      │    adjustment    │    │                  │
      └──────────────────┘    └──────────────────┘
```

Map it back to the mental model: **step 1 is the architect writing the spec, step 2 is the builder laying bricks, steps 3–4 are the architect inspecting the work and ordering the fix.** You stay in the architect's chair the whole loop.

**The trap to avoid:** treating Copilot as a fancy autocomplete. The big productivity gain comes from running the *whole loop*, including the *describe the adjustment* step — many learners stop after step 3 and end up with a brittle blob of generated code they don't fully understand.

> 🎯 **Speed isn't the only point.** Faster loops mean you can *try more variants* before committing. The cost of trying a third architecture sketch went from "half a day" to "thirty seconds". Architecture decisions get better when exploration is cheap.

## 5. Anatomy of a good code prompt — the 5 building blocks

If the prompt is the architect's spec, then a vague spec gets you a generic building. The single biggest predictor of generated-code quality is *prompt specificity*. A vague prompt produces generic code; a specific prompt produces *your* code.

**Every Agent-Mode prompt should include all 5:**

1. **Stack** — name the libraries and frameworks explicitly. *"FastAPI + SQLAlchemy + SQLite"* not *"a web API"*.
2. **Files & folders** — give the layout. *"`backend/main.py`, `backend/models.py`, `frontend/app.py`"* not *"a Python project"*.
3. **Data model & endpoints** — concrete schemas. *"Table `uploads` with columns `id, filename, n_rows, uploaded_at`. Endpoint `POST /api/upload`."*
4. **Demo data path** — *"include a `sample_data.py` that seeds the DB with 500 synthetic rows so the app works on first run."*
5. **Constraints** — CORS, ports, allowed libraries, error behaviour. *"CORS allow `http://localhost:8501`. Pydantic schemas. Standard library only — no external dependencies beyond requirements.txt."*

> 💡 **Rule of thumb:** *bad prompts produce generic code; good prompts produce *your* code.*

> 🔭 These build on the general code-prompting patterns from **NB 45 §3** (give the model what it can't see, ask for one thing, specify constraints, ask for tests). There you learned *what* makes a good code prompt; here you apply it to scaffolding a whole app with Agent Mode.

### Bad prompt vs good prompt — side by side

Same brief — *"build the ChurnScope backend"* — written two ways. Watch how much the architect leaves to chance in the first version.

**Bad** (vague, produces a generic SQLite + Flask sample that has nothing to do with churn):

```
Build me an app that can analyse customer CSVs and do a bit of ML.
```

**Good** (all 5 building blocks — this is recognisably *ChurnScope*):

The agent has almost nothing left to guess — and the output reliably matches what you wanted. That's the spec we'll actually hand to Agent Mode in NB 28; for now, notice that *the precision is the whole point*.

## 6. The Git-first workflow — GitHub before code

A spec and a builder aren't enough — an architect also keeps a versioned record of every revision, so a bad change can always be undone. For software, that record is Git. The surprise is the *order*.

Most tutorials show you Git *after* you've written code (`git init` locally, then push to remote). For POCs, the reverse order is faster and cleaner: **create the GitHub repo first, then clone, then code.**

```
  1. GitHub repo    →  2. Clone     →  3. Open in     →  4. Create files →  5. Commit & push
     (online)            (git clone)    VS Code           in VS Code        (regularly)
```

**Why this order:**

- README, `.gitignore`, and license are generated by GitHub directly — no boilerplate to remember.
- The remote is connected from second one — no follow-up `git remote add origin ...`.
- Issues, branches, and (later) collaborators are ready from minute one.

### What this looks like in practice

```bash
# After creating the repo on github.com:
git clone https://github.com/<user>/<repo>.git
cd <repo>

# Create a fresh virtual environment
python -m venv .venv
source .venv/bin/activate     # macOS/Linux
# .venv\Scripts\activate     # Windows

# ... write code, run, iterate ...

# Commit regularly
git add .
git commit -m "feat: initial Streamlit scaffold"
git push
```

> 💡 **The golden rule:** commit after *every working change*. Five small commits beat one giant one — easier to review, easier to revert, easier to explain.


### Repo structure for a multi-POC course

If you're doing all three POCs from NB 28 in one repo, the recommended layout:

```
poc-repo/
├── .gitignore             # what NEVER gets versioned
├── README.md              # overview of all three POCs
├── poc1_streamlit_basic/
│   ├── .venv/             # local virtual environment (gitignored)
│   ├── app.py             # Streamlit app
│   ├── requirements.txt
│   └── data/              # sample CSVs
├── poc2_streamlit_fastapi_sqlite/
│   ├── .venv/
│   ├── frontend/          # Streamlit UI
│   ├── backend/           # FastAPI + SQLite
│   ├── requirements.txt
│   └── app.db             # SQLite file (gitignored)
└── poc3_xgboost_churn/    # similar layout plus model.pkl
```

**Conventions:**

- **One repo, three subfolders, one per POC** — easy to clone, one issue board.
- **Each POC has its own `.venv` and `requirements.txt`** — no dependency conflicts between POCs.
- **A shared `README.md` at the root** — short description, setup steps, and run command per POC.

> 🎯 **What goes in the repo:** *only* your own code and what someone else needs to reproduce it on a fresh machine — source code, `requirements.txt`, config templates, docs. **Not** `.venv/`, `__pycache__/`, generated `*.db` or `*.pkl`, or any `.env`.


## 7. `.gitignore` and secrets — getting this wrong is expensive

A `.gitignore` file at the repo root tells Git which files to skip. **Without one, your repo bloats, slows, and leaks secrets.**

**Minimal `.gitignore` for a Python POC:**

```
.venv/
__pycache__/
*.db
*.pkl
.env
*.key
.DS_Store
```

**Categories:**

- **Environment / machine-specific:** `.venv/`, `__pycache__/`, `.DS_Store`, `*.pyc`.
- **Generated at runtime:** `*.db`, `*.pkl`, `models/`, anything the code recreates from sources.
- **Secrets:** `.env`, `.env.*`, `*.key`, `*.pem`, anything an API key could live in.

### Secrets discipline

> ⚠️ **Never commit secrets.** API keys, tokens, passwords, real `.env` files, database files with personal data.

**If it happens anyway:**

1. **Rotate the key immediately.** Deleting from Git history is not enough — it's already on someone's clone or in GitHub's mirrors.
2. Notify your security / DPO team if the key gave access to anything material.

**Prevention:** verify `.gitignore` *before your first commit*. The first commit is the one that sets the pattern — get it right.


## 8. ⚠️ The top 6 vibe-coding pitfalls

A good architect knows where buildings usually crack *before* pouring the foundation. These six are where first POCs almost always crack — every learner hits some of them in their first few projects. Knowing them in advance saves hours.

1. **Secrets in the repo** — `.env` with a real API key gets pushed. *Mitigation: `.gitignore` before first commit; rotate immediately if you slipped.*
2. **Missing `.gitignore`** — `.venv/`, `__pycache__/`, `*.db` get committed. Repo bloats, clones are slow, generated files cause merge conflicts. *Mitigation: copy the standard `.gitignore` template (§7) into every repo.*
3. **Wrong Python environment** — Conda vs. venv vs. system Python; the agent installs into one and you run from another. Errors say *"module not found"* even though you definitely installed it. *Mitigation: always activate the same venv before installing and before running; verify with `which python`.*
4. **Port conflicts** — Streamlit's 8501 or FastAPI's 8000 already taken. Server starts but you see a stale page. *Mitigation: read the terminal output every time; if conflicts persist, change the port with `--port 8502`.*
5. **CORS errors** — Streamlit (8501) calls FastAPI (8000), browser blocks the cross-origin request. *Mitigation: add `CORSMiddleware` allow `http://localhost:8501` on the backend (see NB 28 POC 2 prompt).*
6. **Huge undocumented commits** — *"updates"* as message, 200 files changed. Nothing reviewable, nothing revertable. *Mitigation: commit after every working change; the commit message reads as a sentence ("Add retry logic to the LLM client").*

> 💡 **Golden rule:** *Setup first, code second.* Half an hour of fastidious setup saves half a day of mysterious errors.

## 10. 🔬 The structured-output loop — turning TEXT into reliable DATA

Your first POC (§9) just *displayed* an LLM-style result. But the moment a POC has to **do something** with what the model says — store ChurnScope's per-account risk in a database, branch on a churn category, fill a form — you hit the single most important fact about LLMs:

> **A language model returns a STRING. Your program needs an OBJECT (a `dict`/JSON).**

The model does not hand you `{"sentiment": "positive", "score": 0.9}` as a Python dict. It hands you the *text* `'{"sentiment": "positive", "score": 0.9}'` — and sometimes it wraps that text in chatty prose or a ```json fence, or drops a field entirely. So between "the model spoke" and "my code can use it" there is always a small, boring, **non-optional** pipeline:

```text
   PROMPT  ──►  free TEXT  ──►  PARSE  ──►  VALIDATE  ──►  ┌─ ok ─► USE the dict
 (ask for     (a STRING,     (json.loads  (right keys?    │
  JSON only)   maybe messy)   on the JSON  right types?)  └─ bad ─► RETRY (bounded)
                              substring)                            then give up
```

### The mental model

The architect-and-builder picture has a sibling here, and it's just as load-bearing:

> 🧭 **Treat LLM output exactly like untrusted user input.** It is text, not an object. You must *parse* it, *validate* it, and *handle failure* — never assume the next call is well-formed just because the last one was.

This section builds that pipeline from scratch, fully **offline**, with a deterministic mock so you can watch every step. The exact same shape powers every "extract structured data with an LLM" feature you'll build later — including the moment in NB 28+ when ChurnScope asks Claude to turn a wall of support-ticket text into a clean risk record.

### Step 1 — the model returns messy text (a deterministic mock)

To study the pipeline without an API key or network, we use a **mock** that imitates real-world messiness. Real models are non-deterministic and *occasionally* wrap JSON in markdown fences or add a sentence of preamble. Our `fake_llm` reproduces that on purpose: on its **first** call it returns a fenced, chatty blob; on the **second** call it returns clean JSON. That lets us prove the *retry* path deterministically.

> 💡 Why fences happen at all: models are trained on tons of markdown where code lives inside ```json … ``` blocks, so they imitate that formatting even when you asked for "JSON only". Your parser must be robust to it.

````python
import json, re

# A deterministic stand-in for `client.chat(...)`. No network, no API key.
# It mimics two real failure modes, then a clean success — driven by a call counter.
_FAKE_CALLS = {"n": 0}

def fake_llm(prompt: str) -> str:
    """Pretend to be an LLM asked to return JSON. Returns a STRING (never a dict)."""
    _FAKE_CALLS["n"] += 1
    if _FAKE_CALLS["n"] == 1:
        # Messy: chatty preamble + a ```json fence + a MISSING required key ("score").
        return (
            "Sure! Here is the sentiment analysis you asked for:\n"
            "```json\n"
            '{"sentiment": "positive"}\n'
            "```\n"
            "Let me know if you need anything else."
        )
    # Clean: exactly the JSON we asked for, no fence, no prose.
    return '{"sentiment": "positive", "score": 0.92}'

# Reset and peek at the first (messy) response so the problem is concrete:
_FAKE_CALLS["n"] = 0
raw = fake_llm("Classify the sentiment of: 'I love this product!' Return JSON only.")
print(repr(raw))
print("\ntype of what we got back:", type(raw).__name__)   # <- str, NOT dict
````

### Step 2 — PARSE defensively, then VALIDATE the fields

Two defensive moves turn that messy string into a trustworthy dict:

1. **Extract the JSON substring.** Strip a ```json … ``` fence if present, otherwise grab the text from the first `{` to the last `}`. This throws away the chatty prose around the JSON.
2. **`json.loads` + validate.** Parsing can still fail (truncated output, trailing comma). Even when it *succeeds*, the dict might be missing a required key or have the wrong type. **Parsing proves it's JSON; validation proves it's the JSON you need** — they are two different checks.

We raise a clear `ValueError` on any problem so the caller (the retry loop in Step 3) can decide what to do.

```text
   raw string
       │  strip ```json fence / slice {...}
       ▼
   json.loads  ──(not JSON)──►  raise ValueError
       │ ok
       ▼
   check required keys exist  ──(missing)──►  raise ValueError
       │ ok
       ▼
   check types are right      ──(wrong)──►  raise ValueError
       │ ok
       ▼
   clean validated dict ✅
```

````python
def extract_json(text: str) -> str:
    """Pull the JSON substring out of possibly-chatty, possibly-fenced text."""
    # 1) Prefer the contents of a ```json ... ``` (or plain ``` ... ```) fence.
    fence = re.search(r"```(?:json)?\s*(.*?)```", text, re.DOTALL)
    if fence:
        return fence.group(1).strip()
    # 2) Otherwise take from the first '{' to the last '}' — drops surrounding prose.
    start, end = text.find("{"), text.rfind("}")
    if start != -1 and end != -1 and end > start:
        return text[start:end + 1].strip()
    return text.strip()

def parse_and_validate(text: str) -> dict:
    """TEXT -> validated dict, or raise ValueError describing what's wrong."""
    candidate = extract_json(text)

    # PARSE: is it even JSON?
    try:
        data = json.loads(candidate)
    except json.JSONDecodeError as e:
        raise ValueError(f"not valid JSON: {e}") from None

    # VALIDATE: is it the JSON we need? (required keys + types)
    if not isinstance(data, dict):
        raise ValueError(f"expected a JSON object, got {type(data).__name__}")
    required = {"sentiment": str, "score": (int, float)}
    for key, typ in required.items():
        if key not in data:
            raise ValueError(f"missing required key: {key!r}")
        if not isinstance(data[key], typ):
            raise ValueError(f"key {key!r} has wrong type: {type(data[key]).__name__}")
    return data

# The messy first response: fence is stripped, but the "score" key is missing -> rejected.
messy = (
    "Here you go:\n```json\n{\"sentiment\": \"positive\"}\n```"
)
print("extracted substring:", repr(extract_json(messy)))
try:
    parse_and_validate(messy)
except ValueError as e:
    print("rejected (as it should be):", e)

# A clean response validates fine and gives us a real dict we can use:
clean = '{"sentiment": "positive", "score": 0.92}'
ok = parse_and_validate(clean)
print("accepted:", ok, "| score + 0.01 =", ok["score"] + 0.01)   # arithmetic on real data
````

### Step 3 — on failure, RETRY (bounded)

Because the model's output varies, the robust pattern is a **bounded retry loop**: call the model, try to parse+validate, and if it fails, *try again* — up to a small fixed number of attempts. "Bounded" is the load-bearing word: never `while True`, or one stubbornly-malformed response will spin forever and burn tokens.

A production version also **repairs**: it appends the error to the next prompt (*"Your last reply was not valid JSON — the field `score` was missing. Return JSON only."*) so the model can self-correct. We show that as a `repair_hint` below.

```text
 for attempt in 1..MAX:
     text = call_model(prompt + repair_hint)
     try:    return parse_and_validate(text)   ◄─ success: exit immediately
     except: remember the error, build a repair_hint, loop again
 else:       raise — give up cleanly after MAX tries
```

Our `fake_llm` fails on attempt 1 (missing `score`) and succeeds on attempt 2, so the loop below demonstrates a real recovery — not a lucky first try.

````python
def get_structured(prompt: str, max_attempts: int = 3) -> dict:
    """Call the model, parse+validate, RETRY on failure — bounded, with a repair hint."""
    repair_hint = ""
    last_error = None
    for attempt in range(1, max_attempts + 1):
        text = fake_llm(prompt + repair_hint)          # the model returns a STRING
        try:
            data = parse_and_validate(text)            # TEXT -> validated dict
            print(f"attempt {attempt}: ✅ success -> {data}")
            return data
        except ValueError as e:
            last_error = e
            print(f"attempt {attempt}: ⚠️  {e} -> retrying")
            # Repair: tell the model what went wrong so it can fix the NEXT reply.
            repair_hint = (
                f"\n\nYour previous reply was invalid ({e}). "
                f"Return ONLY valid JSON with keys 'sentiment' (string) and 'score' (number)."
            )
    raise RuntimeError(f"no valid output after {max_attempts} attempts; last error: {last_error}")

# Reset the mock so attempt 1 is the messy one, attempt 2 the clean one:
_FAKE_CALLS["n"] = 0
result = get_structured("Classify sentiment of 'I love this!'. Return JSON only.")
print("\nfinal usable dict:", result)
print("branch on it like normal data:",
      "👍 escalate to marketing" if result["sentiment"] == "positive" else "👀 review")
````

### 🧠 The takeaway — four habits for every LLM feature

You just built the loop every "LLM → structured data" feature relies on. The four habits, in order:

| Step | What you do | Why it's non-optional |
|---|---|---|
| 1. **Prompt** | Ask for *JSON only*, name the exact keys | Tilts the model toward parseable output |
| 2. **Parse** | Strip fences / slice `{…}`, then `json.loads` | The model returns a *string*, never an object |
| 3. **Validate** | Check required keys **and** types | Valid JSON ≠ the JSON you need (a field can be missing) |
| 4. **Retry** | Bounded loop + a repair hint | Output is non-deterministic; the next call may be malformed |

> 🎯 **Mental model to keep.** *The model outputs text, not objects — so parse it, validate it, and handle failure.* Treat every LLM response like untrusted input crossing a trust boundary into your program.

> ⚠️ **Two traps.** (1) Never `while True` on retries — always bound the attempts. (2) Don't confuse parsing with validating: `json.loads('{"sentiment":"positive"}')` *succeeds*, yet the result is still unusable because `score` is missing. Both checks must pass.

> 💡 **In real life you'd reach for a library.** Pydantic (NB 25) turns the manual key/type checks into a declarative model and produces the error string for your repair hint automatically; many SDKs also offer a native *JSON mode* or *tool/function calling* that constrains the model to a schema. They are conveniences layered on top of — not replacements for — the parse → validate → retry loop you just wrote by hand.

### Exercise 5 — ⭐⭐ Debug me 🐞 (review-style)

A colleague is mid-vibe-coding session and complains: *"I asked Copilot to add caching to my function. It did, the app runs, but the cache hits aren't working — every call re-computes."* They show you the agent's diff:

```python
def expensive_fetch(user_id):
    @lru_cache(maxsize=128)
    def inner(uid):
        return _api_call(uid)
    return inner(user_id)
```

Find at least two things that are wrong with the agent's solution.

<details>
<summary>💡 <b>Solution / Answer</b></summary>

**Issues:**

1. **`@lru_cache` is defined *inside* the function.** Every call to `expensive_fetch` creates a *new* `inner` function with a *new* cache. The cache never persists across calls, so nothing is ever a "hit". *Fix:* move the decorator and the cached function out to module level, or just decorate `expensive_fetch` itself.

2. **Missing `from functools import lru_cache`.** The code would fail at `NameError` on the first call — the agent hallucinated that the symbol was already imported (Failure mode A from NB 45: hallucinated APIs).

3. *(Bonus)* **No `typed=` consideration.** If `user_id` could be passed as int and str interchangeably, `lru_cache` treats them as different cache keys. For an HTTP-style API this can cause confusing cache misses. *Fix:* normalise to a single type at the function boundary.

**The wider lesson:** Copilot output *looks plausible* — it includes the right decorator, the right argument. The bugs are in the *placement* of the decorator and a missing import. This is exactly why the *review* step of the vibe-coding loop is non-negotiable.
</details>
