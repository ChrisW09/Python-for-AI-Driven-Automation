## 1. Embedding AI in the BPM lifecycle

Business-Process Management (BPM) thinks of an operation as a closed loop of four phases:

```
         ┌────────────────┐         ┌────────────────┐
         │                │         │                │
  ┌────▶ │   ANALYSE      │ ──────▶ │    DESIGN      │ ──────┐
  │      │  the as-is     │         │   the to-be    │       │
  │      │  process       │         │   process      │       │
  │      └────────────────┘         └────────────────┘       │
  │                                                          ▼
  │      ┌────────────────┐         ┌────────────────┐
  │      │   MONITOR      │ ◀────── │   EXECUTE      │
  │      │  outcomes,     │         │  the new       │
  └──────│  drift, cost   │         │  process       │
         │                │         │                │
         └────────────────┘         └────────────────┘
```

An AI feature is rarely *the entire process*. It is almost always **one step inside one phase of the loop**. Naming which step makes everything downstream easier — the integration, the rollback plan, the metrics.

Three patterns that show up over and over:

**Pattern A — AI inside the DESIGN phase.** The AI suggests a better process. Examples: a model that proposes which marketing campaigns to discontinue based on attribution data; an LLM that drafts a new SOP from existing tickets. *Human approves the redesign; AI doesn't ship anything to customers.*

**Pattern B — AI inside the EXECUTE phase.** The AI is doing real work in the live process. Examples: ticket classification at the *intake* step; first-draft response generation; OCR of invoices. *Most of this course's notebooks (NB 27–32) target this pattern.* Highest pay-off, highest deployment risk.

**Pattern C — AI inside the MONITOR phase.** The AI helps the team understand what's happening. Examples: anomaly detection on the daily ops dashboard; LLM-summary of the last 1,000 customer-feedback comments. *Lower stakes, often the easiest to ship first — but easy to under-value.*

> 💡 **Tip.** When a stakeholder pitches an AI project, ask them to name the BPM phase. If they can't, the project isn't yet ready to scope — they're describing a capability without naming where it lands.

## 2. Governance — drafting a RACI for an AI initiative

RACI is the classic accountability framework — for every project deliverable, ask: who is **R**esponsible (does the work), **A**ccountable (signs off), **C**onsulted (their input matters), **I**nformed (told what happened).

What's special about an AI initiative is that **the AI itself plays a role** in the work — and you need to be deliberate about which one. The framework below is one I've seen work in real deployments.

### Example RACI for *"customer-support ticket triage AI"*

| Deliverable / Decision | Support team lead | ML / AI engineer | Platform / SRE | Compliance | The AI system itself |
|---|---|---|---|---|---|
| Pick the BPM step to automate | **A** | C | I | C | — |
| Decide the success metrics | A | **R** | I | C | — |
| Choose the LLM provider | I | **R** | C | **A** | — |
| Build the classifier | I | **R** | C | I | — |
| Operate the runtime (deploy / scale / on-call) | I | C | **R/A** | I | — |
| Classify an *individual* ticket at runtime | I | I | I | I | **R** (with confidence) |
| Decide what to do with a low-confidence ticket | **R/A** | I | I | I | — |
| Monthly accuracy / cost review | **A** | **R** | C | I | — |
| Rollback the AI in a crisis | A | C | **R** | I | — |

**Three things to internalise:**

1. **The AI is `R` (responsible) for individual decisions** — but a human is always `A` (accountable). The AI cannot be blamed; a human is.
2. **Compliance is `A` for the provider choice**, not the support team. Anything that touches customer data has a compliance veto.
3. **Rollback is owned by the platform team**, not by the ML team. ML decides *whether* to revert; the platform team can actually *do* it within minutes.

## 3. POC → MVP → Production — three projects, not three phases

Meridian's working triage code (NB 51) feels like it's "almost done." It isn't — it's at the *start* of the longest stretch. The reason teams underestimate that stretch so badly is a vocabulary confusion. The single biggest confusion in AI project planning is treating these three labels as *milestones in one project*. They are better understood as **three distinct projects with three distinct goals**.

| Phase | Goal | Audience | What 'done' looks like | Common mistake |
|---|---|---|---|---|
| **POC** *(proof of concept)* | Show the AI **can do the task at all** | Decision-makers | A 5-minute demo on a curated dataset; an honest accuracy number | Treating POC accuracy as production accuracy |
| **MVP** *(minimum viable product)* | Validate that **users will actually use it** to do their work | A small set of real users in production | 5–50 real users, real workflow, real data, real metrics for 4–8 weeks | Skipping MVP because the POC looked good |
| **Production** | The system is the **default tool** for the task | The whole user base | SLOs, on-call, runbooks, dashboards, change management, audit trail | Calling MVP "production" when it has no SLOs |

**The three are NOT a checklist** — each one's success criterion is different, and skipping one almost always costs you when you reach the next.

> 🏢 **Meridian, located.** When NB 51 ended, Meridian had passed the *POC* question ("can it work?" — yes, 87% on 500 historical tickets) but had answered *neither* of the other two. The MVP question ("will 10 real agents use it on live tickets for 6 weeks?") and the production question ("can we run it for all 5 pods, 24/7, when the engineer is asleep?") are the entire rest of Meridian's journey — and Case 1 above walks both.

### Why each phase exists

**POC exists to answer: *"is this feasible at all?"***

Should take days, not months. The data is curated, the failure modes are known, the goal is a yes-or-no answer. *If the answer is no, the project ends here cheaply.* The most expensive POCs are the ones that grew into half-built MVPs.

**MVP exists to answer: *"will real users use it on real work?"***

The technical part of an MVP is the easy 30%. The other 70% is: integrating with existing tools, training users, defining the human-in-the-loop process, measuring adoption + accuracy + cost in production conditions. **Most pilots die in the MVP phase** — not because the AI is wrong, but because the integration cost is real and the change-management was skipped.

**Production exists to answer: *"can we run this 24/7 for 3 years?"***

SLOs (service-level objectives), on-call rotations, runbooks for the top 5 failure modes, dashboards the team actually looks at weekly, an audit trail for compliance review, a change-management process for the next version. None of this is glamorous, all of it is what separates a system that *runs* from one that *quietly degrades and gets switched off in 18 months*.


### A realistic POC → MVP → Production timeline

For a *typical* AI-feature project of moderate ambition (a ticket-triage classifier, a RAG-based help bot, an OCR-then-classify invoice pipeline):

```
  T=0          POC starts           — 1 ML engineer
  T+4w         POC demo             — go / no-go review
                                       (50%+ of projects stop here — correctly)

  T+4w         MVP scoping          — engineer + PM + 2 user reps
  T+12w        MVP live with 5 real users
  T+20w        MVP review           — adoption, accuracy, cost data in hand
                                       (another 30% stop here — also correctly)

  T+20w        Production scoping    — engineer + PM + platform team + compliance
  T+32w        Production rollout    — gradual ramp 5% → 25% → 50% → 100%
  T+40w        Post-launch review    — SLOs being met? Cost as predicted?
                                       Adoption beyond the early adopters?
```

**Roughly 9 months** to take a feature from "we have an idea" to "100% of users on it." The naive estimate is usually *3 months* — which is realistic only for the *technical* part and skips integration, change-management, and production hardening. Plan for 9 months; celebrate if it's 6.





> ⚠️ **The most common confusion**: thinking *production* is just *MVP, but more robust*. It isn't — it's *MVP, plus all the boring operational discipline that lets it run when you're not watching it*. The architectural notebook (NB 50) covers the technical parts; the BPM and RACI sections of *this* notebook cover the organisational parts. Both have to happen.

## 4. Three case studies — what real AI initiatives look like

Each case below is a composite of patterns from real deployments. Read each one as a story; the *Lessons* section is the take-away.


### 📚 Case 1 — Customer-support triage at Meridian (our running example), a 400-person B2B SaaS

> 🏢 **This is the Meridian story landing.** Everything the previous three notebooks decided — task = ticket triage, shape = 3-tier app, change strategy = augmentation, built with a carefully-reviewed AI assistant — plays out below as a full POC → MVP → Production arc. Read it as the payoff for the whole module.

**Situation.** ~1,200 support tickets per week across 5 channels (email, chat, in-app, phone callback, social). 25 support agents in two timezones. Average time-to-first-response: 4.5 hours; goal is 1 hour. Founder is pushing for *"AI to help"* — leadership not yet specific.

**Discovery (weeks 0–2).** The team enumerates tasks per the NB 49 framework. The agent's day breaks down as: 25% reading + classifying incoming tickets, 35% drafting responses, 20% looking up policy in the knowledge base, 15% talking to the customer, 5% admin. The high-pattern / low-stake corner contains the first two; the *time-to-first-response* metric is dominated by ticket sitting in the queue *unassigned* — i.e. waiting for a human to look at it and route it.

**POC (weeks 2–6).** One ML engineer builds a classifier (the NB 28 + 19 patterns) that takes ticket text and outputs `{category, urgency, suggested_team, confidence}`. Test set of 500 historical tickets; classifier hits 87% category-accuracy and 91% urgency-accuracy. Demo to the head of support: yes, ship it as an MVP for two pods (10 agents). *The POC answered only one question — "can it work at all?" — and it did.*

**MVP (weeks 6–18).** AI auto-routes tickets to the right team if confidence > 0.8; otherwise routes to the *unassigned* queue as before. Two-pod pilot. Adoption metrics, accuracy metrics, agent satisfaction metrics tracked weekly. *This is the confidence-gate that lets augmentation drift to substitution exactly as NB 49 §4 predicted.*

Time-to-first-response in the two pilot pods drops from 4.5 hours → 1.2 hours (the AI routes within seconds; human agents pick up much faster because their queue is *already filtered*). Re-routing rate ("AI got it wrong") sits at 4–6% — high enough to notice, low enough to live with. Agent satisfaction *increases* (anonymous survey): they spend less time on triage they found tedious. One agent says "I was scared this was about replacing me; instead, I have time to do the harder cases properly." *That sentence is the task-not-job framing of NB 49 §2, validated by the people living it.*

**Production (weeks 18–40).** Roll out to all 5 pods. SLOs defined: response-time, re-routing rate, hourly cost. Two new failure modes appear at scale: (1) a product release in week 24 introduces new vocabulary the classifier hadn't seen; re-routing rate spikes to 12% for 3 days before someone notices. (2) The LLM provider raises prices in week 31 — the team swaps to a cheaper model via the `llm_providers.py` shim within 4 hours; quality holds because the team has the eval harness from NB 32.

**Lessons.**
1. The *measurable* impact was on time-to-first-response, but the *human* impact was the agents spending less time on tedious work. The first won the budget; the second won adoption.
2. The classifier wasn't the hard part. The integration with the ticketing system + the queue routing + the agent UI for overrides were the hard parts.
3. Pricing-change incident vindicated the `llm_providers` shim. Provider lock-in would have been a 2-week scramble.
4. Vocabulary drift after the product release vindicated the monitoring dashboard. *Without it, the team would have noticed via angry-customer escalations.*

### 📚 Case 2 — Invoice processing in a 60-person finance department

**Situation.** ~8,000 supplier invoices per month, 80% PDFs (mix of clean digital + bad scans), 20% other formats. Three accounts-payable clerks spend ~70% of their time on data entry: read invoice → type 18 fields into the ERP. Goal: free up clerk time so the team can take on more strategic work (cost-recovery audits) without adding headcount.

**Discovery (weeks 0–4).** The team realises the project is *not just AI* — it's also document ingestion (email → PDF → S3 → queue) and ERP integration (validated fields → API call → confirmation). The AI is *one component* of a pipeline. They draft an end-to-end pipeline (the NB 50 ML-pipeline pattern) before writing any code.

**POC (weeks 4–10).** Off-the-shelf OCR (Azure Document Intelligence) extracts text + structure. LLM extracts the 18 fields as a JSON object (NB 31 pattern). Validated with Pydantic (the schema-validation pattern from NB 31). Tested on 200 invoices from 2024: 94% of fields extracted correctly; 88% of invoices have all 18 fields right end-to-end. *Demo to the CFO: green light for MVP, with a clear caveat that 12% of invoices will still need human review.*

**MVP (weeks 10–22).** Live on one supplier (the largest, ~30% of volume) for 4 weeks. The system auto-fills the ERP form; a clerk reviews each entry before submitting (no auto-submit). After 2 weeks, the clerks ask to remove the per-entry review for high-confidence extractions (the per-field confidence patterns from NB 31). New rule: auto-submit if every field has confidence > 0.95, otherwise review. Throughput per clerk doubles.

Two unexpected failures: (1) the OCR struggles with a particular supplier whose invoices have a faint watermark — accuracy drops to 60% for that supplier; the team adds a per-supplier dashboard and a manual override. (2) An LLM hallucination puts a wrong invoice number on one form; the supplier rejects the payment, takes 3 days to sort out. The team adds a hard rule: invoice numbers must match the regex on the document; if not, the entry goes to the review queue regardless of confidence.

**Production (weeks 22–36).** Roll out to all suppliers. SLO: average time-to-payment from invoice arrival drops from 12 days to 5 days. Cost per invoice: €0.18 in LLM + OCR + storage, compared to €4.20 in clerk time. The freed-up clerk time is reallocated to a *cost-recovery audit* project — the original strategic goal.

**Lessons.**
1. The AI was 30% of the project. The PDF ingestion, ERP integration, per-supplier monitoring, and human-review UI were the other 70%.
2. The clerks *asked for* the auto-submit feature — they trusted the system after observing it for 2 weeks. **Trust grew from observation, not from training.**
3. The hallucination incident led to a *rule layer* on top of the LLM (invoice-number regex). Rules + LLM is often better than LLM alone, especially for fields with deterministic patterns.
4. The freed-up time was *actually* used for the planned strategic work — because the team had named that intent at the start of the project. Without naming it, the freed time would have quietly absorbed into other work and the ROI story would have been weaker.


### 📚 Case 3 — "AI for sales analytics" — a project that *didn't* ship

**Situation.** A sales-ops team at a mid-sized B2B company wants *"AI to help with sales analytics."* The Head of Sales has seen an LLM demo summarising a database query in English and is convinced this is the next thing. Budget allocated: 6 engineer-months.

**Discovery (weeks 0–4).** The team can't get a specific problem statement. Interviews with sales managers produce a long list of *"it would be nice if AI could…"* items — none of them obviously high-pattern, all of them tangled with internal-tooling limitations that aren't AI problems.

**Almost-POC (weeks 4–14).** An engineer builds a Slack bot that lets sales managers ask natural-language questions about the CRM. The bot generates SQL, runs it, and explains the result. *Looks great in the demo.* In real use, the bot:

- Generates SQL that is technically correct but uses the wrong table (the CRM has 4 customer tables with overlapping semantics).
- Confidently gives answers that contradict the official numbers from finance.
- Is used by 6 people in week 1, 2 in week 4, 0 in week 8.

**Project quietly cancelled at week 18.** Engineer is reassigned. Sales ops continues to use the same dashboards they already had.

**Post-mortem (one of the better things this team did).** The discovery process never produced a tight problem statement. The Head of Sales had a *technology preference*, not a *workflow problem*. The Slack bot tried to be a general analytics interface; general interfaces almost always lose to specialised ones the users already trust (the existing dashboards). Multiple-table semantic ambiguity is a *data modelling* problem; LLMs can't fix it. The 'works in demo, dies in production' pattern is the textbook *pilot-purgatory* from NB 49.

**Lessons.**
1. **Solution looking for a problem.** Pitch starts with *"AI to help with X"* and never produces a specific user-story. Stop the project at the discovery review, not at week 18.
2. **Discovery is allowed to conclude "don't build it."** Saying *no* to a 6-month investment that won't ship is a cheap save; a successful project starts with discovery being able to say *no*.
3. **General-purpose interfaces struggle against specialised tools that already exist.** A natural-language interface to existing data is *not* automatically better than a well-designed dashboard. Users have to *prefer* it for a specific task, repeatedly.
4. **"AI to help with sales analytics"** is the equivalent of *"computers to help with finance."* True but useless. A useful framing names a specific task, a specific user, and a specific current pain. *"Sales managers spend 20 minutes every Monday morning manually checking 5 dashboards before their weekly review; can AI prepare a 1-paragraph summary before 9 AM?"* — that's a project.

## 5. Synthesis — a one-page checklist for an AI initiative

When a stakeholder hands you an AI-project idea, the following 10 questions will tell you within 30 minutes whether the project is ready to start. **Run this checklist before any budget is committed.**

1. **The user.** *Who specifically uses it?* ("the head of sales" is not specific enough — name a role and a workflow.)
2. **The task.** *Which task does it absorb?* ("sales analytics" is too broad — name the task with a verb.)
3. **The BPM phase.** *Which phase of the BPM loop?* (analyse / design / execute / monitor)
4. **The pattern × stake.** *Where does the task sit on the 2×2 from NB 49 §6?*
5. **Today's baseline.** *How does this task get done today, and how much time does it take?* (If nobody knows, that's a discovery red flag.)
6. **Success metric.** *What number will be measurably different in 8 weeks?*
7. **Failure mode.** *What does it look like when the AI gets it wrong, and who notices?*
8. **Rollback.** *Can you switch it off in 5 minutes if it goes wrong?*
9. **Cost model.** *What's the per-call cost, and what's the implied monthly bill at full scale?*
10. **RACI.** *Who is `A` for runtime behaviour, who is `A` for compliance, who is `A` for rollback?*

If you can answer all 10 in clear sentences, you have a real project. If three or more are vague, **the project isn't ready** — go back to discovery, not to coding.

> 🧠 **Mental model.** The 10-question checklist is *a tool for the technical lead to say no constructively.* "Yes" is easy to say and leads to pilot purgatory. "No, here's what we'd need first" is what good technical leadership sounds like.


## 6. Run it as a team — the seminar format

Everything in this notebook is written so you *can* do it alone — but the real exercise (and the module's stated goal) is to run the POC → MVP journey **as an interdisciplinary team**. Here is the format that works for a 4-person team over four weeks; §2's RACI gives the vocabulary for the roles.

| Role | Owns (Accountable) | Typical background |
|---|---|---|
| **Product owner / business analyst** | Ist-analysis, use-case selection, the *"should this exist?"* answer, final pitch | Business / domain |
| **Data scientist** | Data exploration, model or prompt design, evaluation evidence (NB 18 / NB 32 style) | Analytics |
| **Engineer** | The running POC (Module 9 stack — or a Module 7 use case), repo hygiene, demo | Software |
| **Governance & quality owner** | RACI, risk register, data-protection check, the §5 checklist, honest limits slide | Mixed — deliberately |

**The four-week cadence**

1. **Week 1 — Ist-analysis.** Pick a real process (one team member's employer, a university process, or a §4 case as fallback). Deliverable: as-is process sketch (§1's BPM lens) + problem statement + success metric.
2. **Week 2 — Soll-architecture + RACI.** Deliverable: target architecture sketch (NB 50 patterns), a filled RACI (§2), and the data-protection paragraph. The governance owner presents it; the others attack it.
3. **Week 3 — Build the POC.** Deliverable: a running prototype — a Module 9 build (NB 33–36) or a Module 7 use case (NB 23–26) pointed at the team's data. Engineer drives, everyone commits something (NB 51's PR workflow: every merge reviewed by a non-author).
4. **Week 4 — Readiness + pitch.** Deliverable: the §5 one-page checklist filled in honestly (including at least one *red* item), the §3 POC→MVP→Production roadmap, and a 10-minute pitch: 3 minutes demo, 3 minutes evidence, 3 minutes risks, 1 minute ask.

> 💡 **Grading rubric that keeps it honest (for instructors).** 40 % working POC, 25 % evidence quality (did they *measure* anything?), 25 % governance artefacts (RACI + checklist + risks), 10 % pitch. A team that ships a modest POC with honest limits beats a team that demos vapourware — say this out loud in week 1, or you will get vapourware.

> ⚠️ **The classic team failure mode** is everyone doing the engineer's job in week 3 while the Ist-analysis from week 1 quietly rots. The RACI is the fix: one Accountable name per deliverable, and the §5 checklist review is run by the governance owner, *not* by the engineer.

### Exercise 3 — ⭐⭐ POC vs MVP vs Production

A team brings you a 'production-ready ticket-classifier system.' On inspection: it runs on a single engineer's laptop, gets tested on a frozen dataset from January, and has no monitoring. Is this POC, MVP, or production-ready? Defend the call in one paragraph, and name three concrete things missing for the next level.

<details>
<summary>💡 <b>Solution / Answer</b></summary>

**This is a POC, not even an MVP yet.** Production-ready labels three different states: (a) does it work technically (yes — the demo runs), (b) is it integrated into someone's workflow on real data (no — single laptop, frozen test set), (c) is it operable without the original engineer awake (no — no monitoring, no on-call, no SLOs).

**Three concrete missing things for MVP:** (1) integration into the live ticket queue and the agent UI; (2) a small pilot group (5–10 agents) using it on real tickets for 4 weeks; (3) a dashboard tracking accuracy on labelled production tickets (not just the frozen January set).

**Three additional missing things for production:** (4) SLO definitions and alerting; (5) rollback procedure that doesn't require the original engineer; (6) a documented runbook for the top failure modes (provider outage, accuracy degradation, cost spike).
</details>

### Exercise 5 — ⭐⭐ Debug me 🐞 (post-mortem reading)

Read the following 4-line post-mortem and identify at least 3 distinct things the team should change for next time.

> *"We shipped the recommendation model. Accuracy was 87% in testing. After launch, click-through dropped 6%. Nobody knows why. We've rolled back. Next project: try a different model."*

<details>
<summary>💡 <b>Solution / Answer</b></summary>

**What they should change:**

1. **"Nobody knows why"** is the headline finding. They had no monitoring beyond the launch metric — no per-segment, per-page, per-time-of-day breakdown that would let them diagnose. Fix: every model launch ships *with* a diagnostic dashboard, not after.

2. **Click-through dropping by 6% should have triggered an investigation, not a rollback.** A 6% drop *might* be the model being wrong, but it might also be a related product change, a seasonal pattern, or a measurement artefact. Rolling back without diagnosis discards the only opportunity to learn.

3. **"Try a different model"** is a hardware-store reflex — when the hammer doesn't work, try a bigger hammer. The next project should re-examine the *problem* (is recommendation the right intervention?), the *data* (was the test-set representative of the live distribution?), and the *integration* (did the model's output replace something that was working, like an editorial pick?). The model is rarely the bottleneck.

4. (Bonus) **No mention of users.** Were users *surveyed* about the new recommendations? Did anyone watch a real session? Click-through is one data point; user interviews would have surfaced the cause much faster than tweaking the model.
</details>

### Stretch exercise A — ⭐⭐⭐ Write a 1-page project proposal

Pick a real or hypothetical organisation. Write a 1-page proposal for an AI-automation project. It must contain:

<details>
<summary>💡 <b>Solution / Answer</b></summary>

**Example proposal (one page, illustrative — for a mid-sized B2B fintech):**

> **Project: AI-assisted KYC document review**
>
> **Problem.** Our compliance analysts spend ~70% of their week reading customer KYC documents (ID, proof of address, business-registration) and flagging mismatches against our customer-onboarding records. Average time to clear a customer: 6 days. Goal: drop to 2 days without expanding the team.
>
> **BPM phase.** EXECUTE — the AI is acting on each document as it arrives in the live onboarding pipeline.
>
> **Success metric.** Average time-to-clear drops from 6 days to ≤ 3 days for the pilot pod, with a manual-override rate ≤ 10%.
>
> **POC plan (weeks 0–4).** One ML engineer + one compliance analyst (50%). Build an OCR-then-LLM extractor for the 6 most common document types (covers 80% of volume). Test on 200 historical documents. Demo to the head of compliance in week 4.
>
> **Risks + early warnings.**
>  1. *OCR accuracy varies by document type.* Early warning: per-type accuracy in the test set drops below 90% for any of the 6. Mitigation: route the worst-performing type to manual review.
>  2. *Compliance team uncomfortable with automated checks on legally-sensitive material.* Early warning: pilot analyst expresses doubt in week 1 demo. Mitigation: AI is *advisory only* through the entire MVP — no auto-clear, ever.
>  3. *Edge-case documents (foreign IDs, B2B customers) silently drop in accuracy.* Early warning: re-routing rate creeps above 10% in any week. Mitigation: per-segment dashboard from day one, not added later.

**Why this proposal is fundable:** specific user, specific task, specific metric, specific go/no-go demo, named risks with concrete signals. A reviewer can imagine the work happening — and can imagine what *no* looks like in 4 weeks.
</details>

### Stretch exercise B — ⭐⭐⭐ Design the production rollout

Take the proposal from Stretch A (yours or the example). Sketch a *production rollout plan* — what happens between "MVP succeeded" and "100% of users on it." Include: ramp-up percentages, monitoring thresholds, abort conditions, communication plan, and the named owner of the rollback button.

<details>
<summary>💡 <b>Solution / Answer</b></summary>

**Example rollout plan** (continuing the KYC example):

> **Rollout plan — assumes MVP cleared in week 18 with 5 analysts, accuracy ≥ 92%, override-rate ≤ 8%.**
>
> **Ramp.** Week 1: 10% of new customer onboardings auto-routed through the AI (~50/week). Week 3: 25% if metrics are within SLO. Week 6: 50%. Week 10: 100%.
>
> **Monitoring thresholds.** Daily auto-check; alert and pause-ramp if any one is breached:
> – override-rate > 12% for 3 consecutive days,
> – per-type accuracy < 90% in any of the 6 document types,
> – cost-per-customer > €1.20 (vs €0.65 baseline in MVP),
> – p95 latency > 90 s (vs 45 s baseline).
>
> **Abort conditions.** Any single one of:
> – override-rate > 25% in any 24h window → immediate revert to manual-only,
> – any incident where the AI cleared a customer that should not have been (failure → false positive),
> – LLM provider outage > 4 h → fail-over to backup provider; if both down > 8 h, manual processing resumes.
>
> **Communication plan.** Weekly all-hands update for the first 6 weeks; monthly thereafter. Compliance head gets a dashboard with the four monitoring thresholds.
>
> **Owner of the rollback button.** Platform on-call engineer can revert within 5 minutes (single config flag). Decision authority for non-emergency revert: Head of Compliance.

**The skill the exercise builds**: thinking through *what counts as failure* before launching, and *who is empowered to declare it*. Most production AI projects get this wrong by assuming the team that built it will also recognise its failures — they're often the worst-placed to.
</details>

### Stretch exercise C — ⭐⭐⭐ Compare two of the case studies

Compare Case 1 (support triage) and Case 2 (invoice processing) along these dimensions:

- *Architecture pattern* they used (from NB 50).
- *Change strategy* (NB 49 §4).
- *BPM phase* (this notebook §1).
- *What the AI was responsible for vs what humans were accountable for*.
- *The most important non-AI work* the team had to do.

Then identify *one thing each case could learn from the other*.

<details>
<summary>💡 <b>Solution / Answer</b></summary>

**Comparison:**

| | Case 1 (support triage) | Case 2 (invoice processing) |
|---|---|---|
| Architecture | Service-oriented (classifier service + ticket-queue service + agent UI) | End-to-end ML pipeline (ingestion → OCR → LLM extract → validation → ERP) |
| Change strategy | Augmentation → drifted to substitution for high-confidence cases | Substitution from the start (clerk reviews, doesn't author) |
| BPM phase | EXECUTE (live process) | EXECUTE (live process) |

**One thing each could learn from the other:**

- **Case 1 could learn from Case 2's rule layer.** Case 2 added a regex-check on the invoice-number field as a *rule on top of the LLM*. Case 1 could similarly add rule-based pre-routing for tickets that obviously belong to a team (e.g. matching a known account-manager email signature), reducing the volume the LLM has to process and reducing failure-mode exposure.

A confidence-gate up front would have shortened time-to-trust by weeks.

### Stretch exercise D — ⭐⭐⭐ Write a 'this project should stop' memo

Take Case 3 (the sales-analytics project that didn't ship). Imagine it's week 12 and you've just joined as the new technical lead. Write a 1-page memo to leadership recommending the project be stopped. Make the memo:

- Respectful of the original sponsor's intent.
- Specific about what went wrong (not vague "this won't work").
- Honest about what *would* have been needed instead.
- Constructive about what the team should do next (people don't disappear when a project ends).

Writing the *stop memo* is one of the most underrated skills in technical leadership. Practising on a hypothetical case is much cheaper than the first time you have to do it on a real one.

<details>
<summary>💡 <b>Solution / Answer</b></summary>

**Example memo:**

> **To:** [Head of Sales], [CTO]
> **From:** [Technical Lead], Sales-Analytics AI Project
> **Date:** Week 12
> **Re:** Recommendation to stop the project and reallocate the engineer
>
> Thank you for the trust you placed in this project. After two weeks of close work with the team and one week of observing real usage, I'm recommending we stop here. The reasoning below is meant to be useful for the *next* project we attempt in this space, not an indictment of this one.
>
> **What we built.** A Slack bot that lets sales managers ask natural-language questions about CRM data; it generates SQL, runs it, and explains the result.
>
> **What's not working.** Three things, in priority order. (1) The CRM has 4 customer tables with overlapping semantics; the bot frequently uses the wrong one and gives confidently incorrect answers. Fixing this is a data-modelling project, not an AI project — and that data-modelling project would be at least 3 months. (2) Usage has dropped from 6 weekly users in week 1 to 0 in week 8. Direct interviews with the dropped users surface the same pattern: 'I checked the answer against the dashboard; the answer was wrong; I went back to the dashboard.' (3) The existing dashboards already give sales managers what they need; the bot would have to be *meaningfully better* at something to displace them, and currently it isn't.
>
> **What would have been needed.** A specific *task* (not 'sales analytics' broadly) where a natural-language interface beats a dashboard. The most defensible candidate from our user interviews: *'a Monday-morning email summarising the previous week's pipeline movement'* — a one-shot, schedulable task with a single owner. That is a different project; not a rebuild of this one.
>
> **What I recommend.**
> 1. Stop the bot work immediately. Quietly remove it from the Slack workspace by Friday.
> 2. Re-scope: dedicate two weeks to a *discovery* exercise across sales-ops looking for the highest-leverage *narrow* AI use-case (the Monday-summary or another we surface). 1-page proposal back to you by end of week 14.
> 3. The engineer (X) is reassigned to the SDR-prioritisation project, which has a clear problem statement and was on the backburner.
>
> I know stopping feels like a step back. In my experience, projects of this shape that *keep going* cost 2–3× as much before stopping anyway, and burn the team's appetite for the *next* AI project — which is the one we want to succeed.
>
> Happy to discuss live. — [Name]

**What makes this memo work:** acknowledges the sponsor's intent without flattery; specific failure modes with evidence; proposes what *would* be a real project (not just "do nothing"); reassigns the human respectfully; argues from the next project's perspective, not the current one's regrets.
</details>

## 🧠 Key takeaways

- AI features sit at one of four BPM phases — name the phase before you scope the work.
- Governance for AI is RACI plus *the AI as a participant*: it's `R` for individual decisions; a human is always `A`.
- POC / MVP / Production are not phases of one project — they are three projects with three goals. The most common error is calling MVP "production-ready."
- Most successful AI initiatives are 30% AI and 70% integration + change management + monitoring. Plan accordingly.
- A real project survives the 10-question readiness checklist with specific answers. A pitch that can't is *not* ready to start.
- **"Stop the project"** is a deliverable, not a failure. Writing the stop-memo is a senior skill.

> 🧭 **Back to the mental model.** *POC, MVP, and Production are three projects, not three phases.* Notice how the whole notebook hangs off that one distinction: the BPM lens tells you *which step* you're shipping, the RACI tells you *who's accountable* once it's live, and the three case studies are really three answers to the three questions — Case 1 passed all three, Case 2 passed all three with a rule-layer twist, and Case 3 correctly *failed the first one* and stopped. Confuse the projects and you ship an MVP calling itself production; keep them separate and you know exactly which question you're trying to answer this week.
