## 6. Environment check — let's make sure everything works

Run the cell below. It will:

- Confirm your Python version (3.10+ is required, 3.11+ is recommended).
- Confirm the core libraries are importable.
- Print the versions so a forum post about a bug has the right info.

If anything fails, scroll down to the **troubleshooting** section.

````python
import sys, importlib, platform

REQUIRED = {
    "numpy":         "≥ 1.24",
    "pandas":        "≥ 2.0",
    "matplotlib":    "≥ 3.7",
    "sklearn":       "≥ 1.3   (package name: scikit-learn)",
}
OPTIONAL = {
    "requests":      "needed for Notebook 7 (HTTP)",
    "statsmodels":   "needed for Notebook 14 (time series)",
    "pytest":        "needed for Notebook 23 (testing)",
}

def check(pkg, why):
    try:
        m = importlib.import_module(pkg)
        v = getattr(m, "__version__", "?")
        return f"  ✓ {pkg:<14} {v:<8}  ({why})"
    except ImportError:
        return f"  ✗ {pkg:<14} MISSING   ({why})"

print(f"Python    : {sys.version.split()[0]}  ({platform.platform()})")
print(f"Executable: {sys.executable}\n")
print("Required:")
for p, w in REQUIRED.items():
    print(check(p, w))
print("\nOptional (install when you reach those modules):")
for p, w in OPTIONAL.items():
    print(check(p, w))
````

## 10. Troubleshooting — when something doesn't work

| Symptom | Likely cause | Fix |
|---|---|---|
| `ModuleNotFoundError: No module named X` | Library not installed in this kernel | `!pip install X` *in this notebook* (then restart kernel) |
| Cell runs forever | Something hit a network timeout | Edit the cell, add `timeout=8`, restart kernel, re-run |
| `IndentationError` | Mixed tabs and spaces | Pick "Convert indentation to spaces" in your editor |
| Different output from the solution | Different package version | Check `pip show pandas` etc.; usually harmless |
| Notebook 7 fails to fetch data | No internet in your sandbox | The notebook has a *recorded* fallback — re-run, check the warning |

> 💡 **When you're stuck for more than 15 minutes**, copy the full error traceback into a search engine before doing anything else. 95% of beginner errors have been documented.
